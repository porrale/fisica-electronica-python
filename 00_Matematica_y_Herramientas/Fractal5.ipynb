```python
import numpy as np
import matplotlib.pyplot as plt

# Definición de la función que genera el fractal
def julia(c, max_iter=256, threshold=2):
    # Definir el rango de la grilla
    x_range = (-2, 2)
    y_range = (-2, 2)
    resolution = 500
    
    # Crear una grilla de números complejos
    x = np.linspace(x_range[0], x_range[1], resolution)
    y = np.linspace(y_range[0], y_range[1], resolution)
    z = x[:, np.newaxis] + 1j*y[np.newaxis, :]
    
    # Realizar el cálculo iterativo
    for i in range(max_iter):
        z = z**2 + c
        
    # Determinar qué puntos pertenecen al conjunto de Julia
    julia_set = np.abs(z) < threshold
    
    return julia_set

# Definición de la constante c
c = -0.8 + 0.156j

# Generar el fractal
julia_set = julia(c)

# Mostrar el fractal
plt.figure(figsize=(10, 10))
plt.imshow(julia_set.T, cmap='gray', origin='lower')
plt.axis('off')
plt.show()
```

```python
import random
import matplotlib.pyplot as plt

# Definir las ecuaciones de transformación
eq1 = {"prob": 0.01, "matrix": [[0, 0], [0, 0.16]], "vector": [[0], [0]]}
eq2 = {"prob": 0.85, "matrix": [[0.85, 0.04], [-0.04, 0.85]], "vector": [[0], [1.6]]}
eq3 = {"prob": 0.07, "matrix": [[0.2, -0.26], [0.23, 0.22]], "vector": [[0], [1.6]]}
eq4 = {"prob": 0.07, "matrix": [[-0.15, 0.28], [0.26, 0.24]], "vector": [[0], [0.44]]}
eqs = [eq1, eq2, eq3, eq4]

# Definir el punto inicial
x, y = 0, 0

# Generar los puntos del fractal
n_points = 100000
points = []
for i in range(n_points):
    r = random.uniform(0, 1)
    for eq in eqs:
        if r < eq["prob"]:
            x, y = eq["matrix"][0][0]*x + eq["matrix"][0][1]*y + eq["vector"][0][0], \
                   eq["matrix"][1][0]*x + eq["matrix"][1][1]*y + eq["vector"][1][0]
            break
        else:
            r -= eq["prob"]
    points.append([x, y])

# Graficar el fractal
fig, ax = plt.subplots(figsize=(8, 8))
ax.scatter([p[0] for p in points], [p[1] for p in points], s=0.2, color='green')
ax.axis('off')
plt.show()
```
