$$ F(x)=\frac{1}{\sqrt{1-x}}=(1-x)^{-1/2}$$



```python
import sympy as sp 
x = sp.Symbol('x') 
y = (1-x)**(-1/2)

sp.diff(y,x)

```

$$F'(0)= - \frac{1}{2} .(1-x)^{- \frac{3}{2}}.(-1))=\frac{1}{2} (1 - x)^{-\frac{3}{2}}$$

```python
f_prima=sp.diff(y,x)
sp.diff(f_prima,x)
```

$$F''(0)=- \frac{1}{2}.-\frac{1}{3}(1-x)^{-\frac{5}{2}}.(-1)=\frac{3}{4}(1-x)^{- \frac{5}{2}}$$

#Utilizando series McLaurin
Si aproximamos  v²/c²  hasta el segundo orden, podemos usar la serie de Maclaurin para x→0.


```python
from sympy import *
from sympy.plotting import *


import numpy as np

x = symbols('x')
# serie Centrada en 0
x0 = 0
expr = (1-x)**(-1/2)
#print("La expresión es:")
pprint(expr)
#orden del polinomio 
n = 2
serie=expr.series(x, x0, n).removeO()


print("La serie es:")
pprint(serie)
```
