 Fuerza de atracción entre dos cargas eléctricas
<font size="7">
<font size="6">
La fuerza de atracción entre dos cargas eléctricas de signo opuesto se puede calcular utilizando la Ley de Coulomb. Esta ley establece que la magnitud de la fuerza eléctrica entre dos cargas puntuales es directamente proporcional al producto de las magnitudes de las cargas e inversamente proporcional al cuadrado de la distancia entre ellas. Matemáticamente, la fórmula para la fuerza de atracción entre dos cargas es:

<font size="6">
$ F = \frac{k \cdot |q_1 \cdot q_2|}{r^2} $

</font>



Donde:

- F es la fuerza de atracción entre las cargas (en newtons, N)
- k es la constante electrostática, que tiene un valor de \(8.988 \times 10^9 \, \mathrm{N \cdot m^2/C^2}\)
- \(q_1\) y \(q_2\) son las magnitudes de las cargas (en culombios, C)
- r es la distancia entre las cargas (en metros, m)

<font size="6">
Es importante destacar que la fuerza resultante es atractiva cuando las cargas tienen signos opuestos. Si las cargas tuvieran el mismo signo (positivo o negativo), la fuerza sería de repulsión en lugar de atracción.

</font>


```python
# coding: utf-8

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Circle

def E(q, r0, x, y):
    """Retorna el vector de campo eléctrico E=(Ex,Ey) de una carga q en r0"""
    den = np.hypot(x-r0[0], y-r0[1])**3
    return q * (x - r0[0]) / den, q * (y - r0[1]) / den

nx, ny = 64, 64
x = np.linspace(-2, 2, nx)
y = np.linspace(-2, 2, ny)
X, Y = np.meshgrid(x, y)

count = 1
nq = 2**int(count)
charges = []
for i in range(nq):
    q = i%2 * 2 - 1
    charges.append((q, (np.cos(2*np.pi*i/nq), np.sin(2*np.pi*i/nq))))

Ex, Ey = np.zeros((ny, nx)), np.zeros((ny, nx))
for charge in charges:
    ex, ey = E(*charge, x=X, y=Y)
    Ex += ex
    Ey += ey

fig = plt.figure(figsize=(10, 10))
ax = fig.add_subplot(111)

color = 2 * np.log(np.hypot(Ex, Ey))
ax.streamplot(x, y, Ex, Ey, color=color, linewidth=1, cmap=plt.cm.inferno,
              density=2, arrowstyle='->', arrowsize=1.5)

charge_colors = {True: 'red', False: 'black'}
for q, pos in charges:
    ax.add_artist(Circle(pos, 0.05, color=charge_colors[q>0]))

ax.set_xlim(-2,2)
ax.set_ylim(-2,2)
ax.set_aspect('equal')
ax.set_title('Líneas de flujo y cargas eléctricas')
plt.show()


```

Comentarios explicativos del código:

El código comienza importando los módulos necesarios para la graficación: numpy, matplotlib.pyplot y matplotlib.patches.Circle.

A continuación, se define una función E(q, r0, x, y) que retorna el vector de campo eléctrico (Ex, Ey) de una carga q en una posición r0. Esta función utiliza la función np.hypot() para calcular la distancia entre dos puntos en el plano.

Se definen los puntos de los ejes x e y utilizando numpy.linspace().

Se crea un multipolo con nq cargas. En este caso, se utiliza un dipolo (2 cargas). Las cargas se definen como una lista de tuplas, donde cada tupla contiene la carga (q) y su posición (r0).
Se inicializan dos matrices Ex y Ey de ceros con dimensiones (ny, nx) para almacenar los componentes del campo eléctrico.

A continuación, se realiza un bucle for sobre cada carga en la lista charges. Dentro del bucle, se calculan los componentes del campo eléctrico (ex, ey) utilizando la función E() y se actualizan las matrices Ex y Ey sumando estos componentes.

Luego, se crea una figura (fig) con un tamaño de 10x10 pulgadas y se añade un subplot (ax) a la figura.

Se calcula el color de las líneas de flujo utilizando la función np.log() y np.hypot() para obtener la magnitud del campo eléctrico. Estos valores se multiplican por 2 y se utilizan como argumento para el parámetro color en la función ax.streamplot(). Esto permite representar visualmente la intensidad del campo eléctrico.

Se utiliza la función ax.streamplot() para dibujar las líneas de flujo del campo eléctrico en el subplot. Se especifica el arreglo de coordenadas x e y, los componentes del campo eléctrico Ex y Ey, el color de las líneas, el grosor de línea, el mapa de colores (cmap), la densidad de líneas y el estilo de flecha.

Se define un diccionario charge_colors que asigna colores distintos a las cargas positivas y negativas.

Dentro de otro bucle for, se agrega un objeto Circle (círculo) al subplot para representar las cargas. La posición y el color del círculo dependen de la carga (q) y su posición (pos).

Se configuran las etiquetas de los ejes x e y y se establecen los límites y la relación de aspecto del subplot.

Finalmente, se muestra la figura utilizando plt.show().
