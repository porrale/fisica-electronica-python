Resuelve una intefral definida entre los limitees a y b

```python
from scipy.integrate import quad

def integrand(x):
    return x**2

a, b = 0, 1 # límites de integración
result, error = quad(integrand, a, b)

print(f"El resultado de la integral es: {result:.4f}")
print(f"El error estimado es: {error:.4e}")
```

Para calcular una integral por Python, es necesario utilizar una biblioteca matemática como scipy. Aquí hay un ejemplo de cómo calcular la integral de la función sin(x) desde 0 a pi/2:

```python
import scipy.integrate as spi
import numpy as np

# Definir la función a integrar
def f(x):
    return np.sin(x)

#definir limites
a = 0
b = np.pi
# Calcular la integral
integral, error = spi.quad(f, a, b)

print("La integral de sin(x) desde 0 a 2 * pi es:", integral)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

# Definir los datos para la animación
x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

# Definir la figura y el eje
fig, ax = plt.subplots()
ax.set_xlim(0, 2 * np.pi)
ax.set_ylim(-1, 1)

# Definir la línea que se animará
line, = ax.plot([], [], lw=2)

# Función para actualizar la línea en cada iteración de la animación
def update(frame):
    line.set_data(x[:frame], y[:frame])
    return line,

# Crear la animación
ani = FuncAnimation(fig, update, frames=len(x), interval=50, blit=True)

# Mostrar la animación
from IPython.display import HTML
HTML(ani.to_jshtml())

```

```python
import matplotlib.pyplot as plt

# Definir la función a graficar
def f(x):
    return np.sin(x)

# Crear un array de valores x
x = np.linspace(0,  2* np.pi, 100)

# Crear un array de valores y
y = f(x)

# Graficar la función
plt.plot(x, y, label='sin(x)')

# Graficar la región bajo la curva
plt.fill_between(x, y, 0, alpha=0.3)

# Agregar etiquetas y leyenda
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid()
# Mostrar la gráfica
plt.show()
```
