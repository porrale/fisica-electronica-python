```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import numpy as np

def dibujar_difraccion(ax, tipo):
    ax.set_xlim(0, 10)
    ax.set_ylim(-5, 5)
    ax.set_aspect('equal')
    ax.axis('off')

    # Parámetros de separación lambda para líneas verticales
    if tipo == 'poca':
        lambdas = np.linspace(0.5, 9, 20)
    elif tipo == 'media':
        lambdas = np.linspace(0.5, 9, 10)
    elif tipo == 'mucha':
        lambdas = np.linspace(0.5, 9, 5)

    # Líneas verticales indicando separación entre frentes de onda (en celeste)
    for x_l in lambdas:
        ax.plot([x_l, x_l], [-5, 5], color='deepskyblue', linestyle='--', linewidth=0.5)

    # Líneas planas horizontales (ondas incidentes) en naranja
    for y in np.linspace(-4, 4, 9):
        ax.plot([0, 4], [y, y], color='orange')
        ax.arrow(3.8, y, 0.1, 0, head_width=0.2, head_length=0.2, fc='orange', ec='orange')

    # Pared con rendija
    rendija_ancho = {
        'poca': 2.5,
        'media': 1.5,
        'mucha': 0.8
    }[tipo]

    espacio = rendija_ancho / 2
    ax.add_patch(patches.Rectangle((4, espacio), 0.5, 5 - espacio, color='gray'))
    ax.add_patch(patches.Rectangle((4, -5), 0.5, 5 - espacio, color='gray'))

    # Ondas difractadas según el tipo
    if tipo == 'poca':
        for y in np.linspace(-1, 1, 3):
            ax.plot([4.5, 9], [y, y], color='orange')
            ax.arrow(8.5, y, 0.3, 0, head_width=0.2, head_length=0.2, fc='orange', ec='orange')

    elif tipo == 'media':
        for r in [1.5, 3, 4.5]:
            arco = patches.Arc((4.5, 0), r*2, r*2, theta1=-60, theta2=60, color='orange')
            ax.add_patch(arco)
        for ang in np.linspace(-60, 60, 7):
            rad = np.deg2rad(ang)
            ax.arrow(4.5, 0, 3*np.cos(rad), 3*np.sin(rad),
                     head_width=0.2, head_length=0.2, fc='orange', ec='orange')

    elif tipo == 'mucha':
        for r in [1.5, 3, 4.5]:
            arco = patches.Arc((4.5, 0), r*2, r*2, theta1=-90, theta2=90, color='orange')
            ax.add_patch(arco)
        for ang in np.linspace(-90, 90, 9):
            rad = np.deg2rad(ang)
            ax.arrow(4.5, 0, 3*np.cos(rad), 3*np.sin(rad),
                     head_width=0.2, head_length=0.2, fc='orange', ec='orange')

# Crear figura
fig, axs = plt.subplots(1, 3, figsize=(18, 5))

dibujar_difraccion(axs[0], 'poca')
axs[0].text(6.5, -4.7, r'$\lambda \ll d$', fontsize=14, ha='center')

dibujar_difraccion(axs[1], 'media')
axs[1].text(6.5, -4.7, r'$\lambda \approx d$', fontsize=14, ha='center')

dibujar_difraccion(axs[2], 'mucha')
axs[2].text(6.5, -4.7, r'$\lambda \gg d$', fontsize=14, ha='center')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import numpy as np

def dibujar_difraccion(ax, tipo):
    ax.set_xlim(0, 10)
    ax.set_ylim(-5, 5)
    ax.set_aspect('equal')
    ax.axis('off')

    # Definir color naranja más oscuro
    naranja_oscuro = '#cc6600'

    # Ancho de rendija por tipo
    rendija_ancho = {
        'poca': 2.5,
        'media': 1.5,
        'mucha': 0.8
    }[tipo]
    espacio = rendija_ancho / 2

    # Separación lambda
    if tipo == 'poca':
        dx = 0.4  # más chico para líneas más juntas
    elif tipo == 'media':
        dx = 1
    elif tipo == 'mucha':
        dx = 2

    # Líneas verticales indicando separación lambda
    x_vals = np.arange(0.5, 9, dx)
    for x_l in x_vals:
        if tipo == 'poca' and x_l > 4.5:
            ax.plot([x_l, x_l], [-espacio, espacio], color='deepskyblue', linestyle='--', linewidth=1)
        else:
            ax.plot([x_l, x_l], [-5, 5], color='deepskyblue', linestyle='--', linewidth=0.5)

    # Líneas horizontales (ondas incidentes) en naranja oscuro
    for y in np.linspace(-4, 4, 9):
        ax.plot([0, 4], [y, y], color=naranja_oscuro)
        ax.arrow(3.8, y, 0.1, 0, head_width=0.2, head_length=0.2, fc=naranja_oscuro, ec=naranja_oscuro)

    # Pared con rendija
    ax.add_patch(patches.Rectangle((4, espacio), 0.5, 5 - espacio, color='gray'))
    ax.add_patch(patches.Rectangle((4, -5), 0.5, 5 - espacio, color='gray'))

    # Ondas difractadas
    if tipo == 'poca':
        y_vals = np.linspace(-rendija_ancho/2 + 0.2, rendija_ancho/2 - 0.2, 3)
        for y in y_vals:
            ax.plot([4.5, 9], [y, y], color=naranja_oscuro)
            ax.arrow(8.5, y, 0.3, 0, head_width=0.2, head_length=0.2, fc=naranja_oscuro, ec=naranja_oscuro)

    elif tipo == 'media':
        for r in [1.5, 3, 4.5]:
            arco = patches.Arc((4.5, 0), r*2, r*2, theta1=-60, theta2=60, color=naranja_oscuro)
            ax.add_patch(arco)
        for ang in np.linspace(-60, 60, 7):
            rad = np.deg2rad(ang)
            ax.arrow(4.5, 0, 3*np.cos(rad), 3*np.sin(rad),
                     head_width=0.2, head_length=0.2, fc=naranja_oscuro, ec=naranja_oscuro)

    elif tipo == 'mucha':
        for r in [1.5, 3, 4.5]:
            arco = patches.Arc((4.5, 0), r*2, r*2, theta1=-90, theta2=90, color=naranja_oscuro)
            ax.add_patch(arco)
        for ang in np.linspace(-90, 90, 9):
            rad = np.deg2rad(ang)
            ax.arrow(4.5, 0, 3*np.cos(rad), 3*np.sin(rad),
                     head_width=0.2, head_length=0.2, fc=naranja_oscuro, ec=naranja_oscuro)

# Crear figura
fig, axs = plt.subplots(1, 3, figsize=(18, 5))

dibujar_difraccion(axs[0], 'poca')
axs[0].text(7, -4.7, r'$\lambda \ll d$', fontsize=14, ha='center')

dibujar_difraccion(axs[1], 'media')
axs[1].text(7, -4.7, r'$\lambda \approx d$', fontsize=14, ha='center')

dibujar_difraccion(axs[2], 'mucha')
axs[2].text(7, -4.7, r'$\lambda \gg d$', fontsize=14, ha='center')

plt.tight_layout()
plt.show()
```

# Interferencia por Doble Rendija – Representación Gráfica con Python

Esta visualización muestra dos representaciones del experimento clásico de interferencia:

### (a) Rayos que van desde dos rendijas hasta un punto P sobre una pantalla
- Dos ondas parten desde \( S_1 \) y \( S_2 \)
- Recorren distancias \( r_1 \) y \( r_2 \) hasta el punto \( P \)
- La diferencia en el recorrido produce interferencia
- Se indica el ángulo \( \theta \), la distancia entre rendijas \( d \), y la distancia a la pantalla \( D \)

### (b) Representación geométrica de la diferencia de caminos
- Se muestra que la diferencia de camino es \( \Delta l = d \sin(\theta) \)
- Se construye el triángulo rectángulo con cateto opuesto \( \Delta l \)
- Se repite el ángulo \( \theta \) para mostrar que los rayos son paralelos

Ambas figuras ayudan a visualizar la base del patrón de interferencia observado en pantalla.


```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import numpy as np

# Crear figura
fig, ax = plt.subplots(figsize=(10, 7))

# ---------- Parámetros ----------
rendija_centro1 = (2, 4.0)
rendija_centro2 = (2, 2.5)
rendija_altura = 0.5
barrera_ancho = 0.2
barrera_color = "gray"
r_max = 9
n_arcos = 9
separacion = 0.5

# ---------- Barrera con rendijas vacías ----------
# Se dibujan tres secciones de barrera (arriba, entre y abajo de las rendijas)
barrera_top = patches.Rectangle((2 - barrera_ancho / 2, rendija_centro1[1] + rendija_altura / 2),
                                barrera_ancho, 1.0, color=barrera_color)
barrera_mid = patches.Rectangle((2 - barrera_ancho / 2, rendija_centro2[1] + rendija_altura / 2),
                                barrera_ancho, rendija_centro1[1] - rendija_centro2[1] - rendija_altura,
                                color=barrera_color)
barrera_bot = patches.Rectangle((2 - barrera_ancho / 2, rendija_centro2[1] - rendija_altura / 2 - 1.0),
                                barrera_ancho, 1.0, color=barrera_color)
ax.add_patch(barrera_top)
ax.add_patch(barrera_mid)
ax.add_patch(barrera_bot)

# ---------- Ondas circulares desde los centros de las rendijas ----------
for i in range(1, n_arcos + 1):
    r = i * separacion
    for center in [rendija_centro1, rendija_centro2]:
        arc = patches.Arc(center, width=2*r, height=2*r, angle=0,
                          theta1=-30, theta2=30, color="deepskyblue", lw=1)
        ax.add_patch(arc)

# ---------- Rayos hacia la pantalla ----------
theta_rays = np.linspace(-30, 30, 9)
for angle in theta_rays:
    for center in [rendija_centro1, rendija_centro2]:
        x_end = 10
        rad = np.deg2rad(angle)
        y_end = center[1] + np.tan(rad)*(x_end - center[0])
        ax.annotate('', xy=(x_end, y_end), xytext=center,
                    arrowprops=dict(arrowstyle="->", color="#cc6600", lw=1.5))

# ---------- Pantalla ----------
ax.plot([10, 10], [0, 7], color="gray", linewidth=5)
for i, y in enumerate(np.linspace(0.5, 6.5, 9)):
    ax.text(10.2, y, "máx" if i % 2 == 0 else "mín", fontsize=10, va='center')

# ---------- Etiquetas ----------
ax.text(2.1, rendija_centro1[1], r"$S_1$", fontsize=12)
ax.text(2.1, rendija_centro2[1], r"$S_2$", fontsize=12)
ax.text(1.5, 3.5, "Barrera", rotation=90, fontsize=10)
ax.text(10.3, 6.8, "Pantalla", fontsize=10, rotation=90)

# ---------- Estética ----------
ax.set_xlim(0, 12)
ax.set_ylim(0, 7)
ax.set_aspect('equal')
ax.axis('off')
ax.set_title("Interferencia de ondas desde doble rendija", fontsize=14)

plt.show()
```
