## 🌊 Difracción en Una Sola Rendija

Cuando la luz pasa por una **rendija estrecha**, se dispersa formando un patrón característico:

$
I(\theta) \propto \text{sinc}^2\!\left(\frac{\pi a \sin\theta}{\lambda}\right)
$

donde:
- $ \lambda $: longitud de onda
- $ a $: ancho de la rendija
- $ \theta $: ángulo respecto al eje central

El resultado es una **franja central brillante** mucho más intensa y ancha, seguida de franjas laterales cada vez más débiles.


```python
# Patrón de difracción de una sola rendija
import numpy as np
import matplotlib.pyplot as plt

# Parámetros físicos
lambda_light = 500e-9  # longitud de onda (500 nm)
a = 1e-4              # ancho de la rendija (0.1 mm)
L = 1.0               # distancia a la pantalla (1 m)

# Ángulo máximo y puntos
theta = np.linspace(-0.02, 0.02, 2000)  # radianes

# Factor de difracción
alpha = (np.pi * a * np.sin(theta)) / lambda_light
intensity = (np.sinc(alpha/np.pi))**2

# Posición en la pantalla
x = L * np.tan(theta) * 1000  # mm

# Graficar
plt.figure(figsize=(10, 6))
plt.plot(x, intensity, label='Difracción (rendija única)')
plt.title('Patrón de Difracción de Una Sola Rendija')
plt.xlabel('Posición en la pantalla (mm)')
plt.ylabel('Intensidad relativa')
plt.grid(True)
plt.legend()
plt.show()
```
