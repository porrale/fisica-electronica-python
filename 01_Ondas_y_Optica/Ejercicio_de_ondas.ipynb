# Patrón de difusión de una estación de radio
## Enunciado:
Una estación de radio que opera en la frecuencia de $1500\text{ kHz} = 1.5 \times 10^6\text{ Hz}$ tiene dos antenas dipolares verticales idénticas que oscilan en fase y que se encuentran separadas por una distancia de $400\text{ m}$.

A distancias mucho mayores que $400\text{ m}$, ¿en qué direcciones es máxima la intensidad en el patrón de radiación resultante?(Nota: Este problema se puede modelar utilizando las ideas de la interferencia de dos ranuras o fuentes puntuales).


 Éste no es un problema hipo￾tético, ya que a menudo es deseable transmitir la mayor parte de la
energía irradiada desde una transmisora de radio en ciertas direcciones
y no de modo uniforme en todas direcciones. No es raro que se utili￾cen pares o filas de antenas para producir el patrón de radiación que
se desea.

Explicación Paso a Paso1. Identificación de variablesPara resolver este problema, primero debemos determinar la longitud de onda ($\lambda$) de la señal de radio. Usamos la relación entre la velocidad de la luz ($c \approx 3 \times 10^8\text{ m/s}$), la frecuencia ($f$) y la longitud de onda:$$\lambda = \frac{c}{f}$$2. Condición de Interferencia Constructiva (Máximos)Para fuentes que oscilan en fase, los máximos de intensidad ocurren cuando la diferencia de trayectoria es un número entero de longitudes de onda ($m\lambda$). La fórmula para los ángulos $\theta$ de estos máximos es:$$d \cdot \sin(\theta) = m\lambda \implies \sin(\theta) = \frac{m\lambda}{d}$$Donde:$d = 400\text{ m}$ (distancia entre antenas).$m = 0, \pm 1, \pm 2, \dots$ (orden del máximo).3.



 Determinación de ángulos válidosComo el valor de $\sin(\theta)$ debe estar entre $-1$ y $1$, solo los valores de $m = 0, \pm 1, \pm 2$ son físicamente posibles:

  Interferencia Destructiva (Mínimos)
  Para que la intensidad sea mínima (puntos de silencio o cancelación), las ondas que llegan de ambas antenas deben estar en oposición de fase. Esto ocurre cuando la diferencia de trayectoria es un número medio entero de longitudes de onda.La fórmula para los ángulos de los mínimos es:$$d \cdot \sin(\theta) = \left( m + \frac{1}{2} \right) \lambda$$O despejando el ángulo:$$\sin(\theta) = \frac{(m + 1/2)\lambda}{d}$$




```python
import numpy as np

# Definición de constantes
c = 3e8  # Velocidad de la luz en m/s
f = 1.5e6  # Frecuencia en Hz (1500 kHz)
d = 400  # Distancia entre antenas en metros

# 1. Calcular longitud de onda (lambda)
wavelength = c / f
print(f"Longitud de onda (λ): {wavelength} m")

# 2. Encontrar los órdenes m posibles
# El valor máximo de m ocurre cuando sin(theta) = 1, entonces m_max = d / lambda
m_max = int(d / wavelength)
m_values = np.arange(-m_max, m_max + 1)

print(f"Órdenes de m válidos: {m_values}")

# 3. Calcular los ángulos para los máximos de intensidad
print("\nDirecciones de máxima intensidad (Interferencia Constructiva):")
for m in m_values:
    sin_theta = (m * wavelength) / d
    # Usamos arcsin para obtener el ángulo en radianes y luego convertimos a grados
    theta_rad = np.arcsin(sin_theta)
    theta_deg = np.rad2deg(theta_rad)
    print(f"Orden m = {m:2d}: θ = {theta_deg:6.1f}°")

# 4. Cálculo de mínimos (Opcional, según la evaluación del ejercicio)
print("\nDirecciones de mínima intensidad (Interferencia Destructiva):")
# La condición es sin(theta) = (m + 0.5) * lambda / d
for m in np.array([-2, -1, 0, 1]):
    sin_theta_min = (m + 0.5) * wavelength / d
    theta_min_deg = np.rad2deg(np.arcsin(sin_theta_min))
    print(f"Mínimo cerca de m = {m:2.1f}: θ = {theta_min_deg:6.1f}°")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parámetros del sistema
d = 400          # Distancia entre antenas (m)
f = 1.5e6        # Frecuencia (Hz)
c = 3e8          # Velocidad de la luz (m/s)
lam = c / f      # Longitud de onda (200m)

# Creamos un rango de ángulos de -90 a 90 grados
theta = np.linspace(-np.pi/2, np.pi/2, 1000)

# Fórmula de la intensidad relativa para dos fuentes en fase:
# I = I0 * cos^2(phi/2) donde phi = (2 * pi * d * sin(theta)) / lambda
phi = (2 * np.pi * d * np.sin(theta)) / lam
intensidad = np.cos(phi / 2)**2





# --- GRÁFICO Patrón Polar (Distribución espacial) ---
plt.figure(figsize=(8, 8))
ax = plt.subplot(111, projection='polar')
# Ajustamos el ángulo para que 0° sea el eje vertical como en el diagrama del libro
ax.set_theta_zero_location("N")
ax.plot(theta, intensidad, color='purple', lw=2)
ax.set_title("Distribución de Energía (Vista Aérea)", va='bottom')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parámetros técnicos del sistema
d = 100          # Distancia entre antenas (m)
f = 1.5e6        # Frecuencia (Hz)
c = 3e8          # Velocidad de la luz (m/s)
lam = c / f      # Longitud de onda (200m)

# Creamos un rango completo de ángulos (0 a 360 grados)
# Usamos 1000 puntos para que la curva se vea suave
theta = np.linspace(0, 2 * np.pi, 1000)

# Cálculo de la intensidad relativa
# La diferencia de fase phi depende del ángulo theta
# phi = (2 * pi * d / lambda) * sin(theta)
phi = (2 * np.pi * d * np.sin(theta)) / lam
intensidad = np.cos(phi / 2)**2

# Configuración del gráfico polar
plt.figure(figsize=(8, 8))
ax = plt.subplot(111, projection='polar')

# Ajustes para que coincida con la orientación del libro:
# 0° arriba (Norte), y sentido horario
ax.set_theta_zero_location("N")
ax.set_theta_direction(-1)

# Dibujar el patrón
ax.plot(theta, intensidad, color='teal', linewidth=2.5, label='Intensidad de señal')
ax.fill(theta, intensidad, color='teal', alpha=0.2) # Relleno para mejor visualización

# Añadir títulos y etiquetas
ax.set_title("Patrón de Radiación de la Estación de Radio\n(Vista Aérea / Difracción)",
             va='bottom', fontsize=14, fontweight='bold')
ax.set_rticks([0.25, 0.5, 0.75, 1.0])  # Escala de intensidad
ax.set_yticklabels(["25%", "50%", "75%", "Máx"])
ax.grid(True, linestyle='--', alpha=0.7)

plt.legend(loc='upper right', bbox_to_anchor=(1.2, 1.1))
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import Arc

# Configuración de la figura - Aumentamos el tamaño para ver bien las ondas
fig, ax = plt.subplots(figsize=(12, 8))
ax.set_aspect('equal')

# Parámetros
d = 400
s1_pos = -d/2
s2_pos = d/2
num_waves = 10
radius_step = 200 # Esta es nuestra longitud de onda (lambda)

# 1. Dibujar las ondas
for r in range(1, num_waves + 1):
    radius = r * radius_step
    circle1 = plt.Circle((s1_pos, 0), radius, color='cornflowerblue', fill=False, alpha=0.4, linewidth=1.5)
    circle2 = plt.Circle((s2_pos, 0), radius, color='cornflowerblue', fill=False, alpha=0.4, linewidth=1.5)
    ax.add_patch(circle1)
    ax.add_patch(circle2)

# 2. Fuentes S1 y S2
ax.plot([s1_pos, s2_pos], [0, 0], 'k-', linewidth=1, zorder=3)
ax.scatter([s1_pos, s2_pos], [0, 0], color='#1f77b4', s=150, zorder=5, edgecolor='black')
ax.text(s1_pos - 80, -100, '$S_1$', fontsize=14, fontweight='bold')
ax.text(s2_pos + 30, -100, '$S_2$', fontsize=14, fontweight='bold')

# 3. Líneas de interferencia (Flechas púrpuras)
# 60° y 120° en coordenadas polares estándar = 30° respecto a la vertical
angles = [60, 90, 120]
arrow_length = radius_step * num_waves # Para que lleguen al borde de la última onda

for ang in angles:
    rad = np.radians(ang)
    dx = arrow_length * np.cos(rad)
    dy = arrow_length * np.sin(rad)
    # Las flechas deben salir del origen (0,0) para pasar por las intersecciones
    ax.annotate("", xy=(dx, dy), xytext=(0, 0),
                arrowprops=dict(arrowstyle="->", color="purple", lw=2.5, zorder=10))

# 4. Línea de base horizontal extendida
ax.annotate("", xy=(arrow_length, 0), xytext=(-arrow_length, 0),
            arrowprops=dict(arrowstyle="<->", color="purple", lw=2))

# 5. Arcos de 30 grados
# El arco se dibuja indicando el ángulo desde la vertical (90°)
ax.add_patch(Arc((0, 0), 500, 500, theta1=60, theta2=90, color='black', lw=1.5))
ax.add_patch(Arc((0, 0), 500, 500, theta1=90, theta2=120, color='black', lw=1.5))
ax.text(80, 280, '$30^\circ$', fontsize=13)
ax.text(-150, 280, '$30^\circ$', fontsize=13)

# 6. Cota de 400 m
ax.annotate('', xy=(s1_pos, -120), xytext=(s2_pos, -120),
            arrowprops=dict(arrowstyle='<->', color='black'))
ax.text(-50, -180, '400 m', fontsize=12)

# Ajustes de visualización dinámicos basados en el número de ondas
limit = arrow_length + 100
ax.set_xlim(-limit, limit)
ax.set_ylim(-250, limit)
ax.axis('off')

plt.title("Interferencia: d=400m, $\lambda$=200m", fontsize=16)
plt.show()
```
