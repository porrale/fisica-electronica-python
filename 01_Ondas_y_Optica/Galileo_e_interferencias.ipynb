```python
import matplotlib.pyplot as plt

# Parámetros del problema
v_x = 10  # Velocidad del objeto y del observador en el eje x
v_y = 20  # Velocidad inicial del objeto en el eje y
g = 9.81  # Aceleración de la gravedad
t_max = 4  # Tiempo máximo de simulación

# Resolución de la caída libre en el eje y
t = 0
dt = 0.01
y = [0]
v = [v_y]
while t < t_max:
    y.append(y[-1] + v[-1]*dt)
    v.append(v[-1] - g*dt)
    t += dt

# Transformación al sistema de referencia del observador
x_obs = [i*v_x for i in range(len(y))]
y_obs = y

# Transformación al sistema de referencia original
x = [x_obs[i] + v_x*t for i in range(len(y_obs))]
y = y_obs

# Gráfico del recorrido del objeto
fig, ax = plt.subplots()
ax.plot(x, y)
ax.set_xlabel('Posición en x')
ax.set_ylabel('Posición en y')
ax.set_title('Recorrido del objeto')
plt.show()
```

Este código creará una animación que muestra la trayectoria de un objeto que se mueve a velocidad constante y luego es lanzado hacia arriba en el eje y. La velocidad inicial del objeto se establece en v = 10 metros/segundo y la duración total de la animación se establece en total_time = 4 segundos. La aceleración debida a la gravedad se establece en g = 9.8 metros/segundo^2 y el intervalo de tiempo se establece en delta_t = 0.01 segundos.

La animación se puede personalizar cambiando los valores de v, total_time, g y delta_t para ajustar la velocidad inicial del objeto, la duración de la animación, la aceleración debida a la gravedad y el intervalo de tiempo, respectivamente. También puedes cambiar la posición inicial del objeto estableciendo los valores de x y y.

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# Definir la velocidad del objeto y el tiempo total de la animación
v = 10  # metros/segundo
total_time = 4  # segundos

# Definir la posición inicial del objeto y el tiempo inicial
x = 0  # metros
y = 0  # metros
t = 0  # segundos

# Definir la aceleración debida a la gravedad y el intervalo de tiempo
g = 9.8  # metros/segundo^2
delta_t = 0.01  # segundos

# Crear la figura y el eje
fig, ax = plt.subplots()

# Establecer los límites de los ejes
ax.set_xlim(0, v*total_time)
ax.set_ylim(0, 2*v**2/g)

# Crear una línea para la trayectoria del objeto
line, = ax.plot([], [], 'o-', label='Trayectoria')

# Agregar una leyenda
ax.legend()
plt.grid()
# Función para actualizar la animación en cada fotograma
def update(frame):
    global x, y, t
    
    # Actualizar la posición del objeto a lo largo del eje x
    x = v * t
    
    # Actualizar la posición del objeto a lo largo del eje y
    y = v * t - 0.5 * g * t**2
    
    # Actualizar el tiempo
    t += delta_t
    
    # Actualizar la línea de la trayectoria del objeto
    line.set_data(x, y)
    
    return line,

# Crear la animación
ani = FuncAnimation(fig, update, frames=np.arange(0, total_time*5, delta_t*10), interval=delta_t*1000)

# Mostrar la animación en el notebook
HTML(ani.to_jshtml())
```

```python
import matplotlib.pyplot as plt
import matplotlib.animation as animation

# Parámetros del problema
v_x = 10  # Velocidad del objeto y del observador en el eje x
v_y = 20  # Velocidad inicial del objeto en el eje y
g = 9.81  # Aceleración de la gravedad
t_max = 4  # Tiempo máximo de simulación

# Resolución de la caída libre en el eje y
t = 0
dt = 0.01
y = [0]
v = [v_y]
while t < t_max:
    y.append(y[-1] + v[-1]*dt)
    v.append(v[-1] - g*dt)
    t += dt

# Creación de la animación
fig, ax = plt.subplots()
line, = ax.plot([], [])
point, = ax.plot([], [], 'o')
ax.set_xlim(1, t_max * v_x/2.1)

ax.set_ylim(0, max(y))

def update(frame):
    # Transformación al sistema de referencia del observador
    x_obs = [i*v_x - frame*v_x for i in range(len(y))]
    y_obs = y

    # Actualización de la posición del objeto en la animación
    #line.set_data(x_obs, y_obs)
    point.set_data(x_obs, y_obs)
    return point,

ani = animation.FuncAnimation(fig, update, frames=int(t_max/dt), interval=10, blit=True)
plt.grid()
plt.close()

# Visualización de la animación en Google Colab
from IPython.display import HTML
HTML(ani.to_jshtml())
```

La difracción es un fenómeno ondulatorio que ocurre cuando una onda encuentra un obstáculo o pasa por una abertura. La onda se curva alrededor del objeto y se dispersa, creando un patrón de difracción en el espacio más allá del obstáculo o la abertura. Este patrón de difracción puede ser observado cuando la onda es medida o visualizada.

La interferencia es otro fenómeno ondulatorio en el que dos o más ondas se superponen y crean un patrón de interferencia. Si las ondas están en fase (es decir, sus crestas y valles se alinean), se produce una interferencia constructiva y el resultado es una amplitud mayor. Si las ondas están fuera de fase (es decir, sus crestas y valles no se alinean), se produce una interferencia destructiva y el resultado es una amplitud menor. El patrón de interferencia resultante puede ser observado cuando las ondas se combinan y se miden o visualizan.

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, FloatSlider

def graficar_ondas(amplitud_1, long_onda_1, fase_1, amplitud_2, long_onda_2, fase_2):
    # Crear eje x
    x = np.linspace(0, 1, 1000)

    # Calcular ondas
    y_1 = amplitud_1 * np.sin(2 * np.pi * x / long_onda_1 + fase_1)
    y_2 = amplitud_2 * np.sin(2 * np.pi * x / long_onda_2 + fase_2)

    # Calcular suma de ondas
    y_suma = y_1 + y_2

    # Graficar ondas
    plt.plot(x, y_1, label='Onda 1')
    plt.plot(x, y_2, label='Onda 2')
    plt.plot(x, y_suma, label='Suma')

    # Agregar leyenda
    plt.legend()

# Crear sliders para gráfico interactivo
grafico_interactivo = interactive(graficar_ondas,
                                  amplitud_1=FloatSlider(min=0, max=1, value=0.5),
                                  long_onda_1=FloatSlider(min=0.1, max=1, value=0.5),
                                  fase_1=FloatSlider(min=0, max=2*np.pi, value=0),
                                  amplitud_2=FloatSlider(min=0, max=1, value=0.5),
                                  long_onda_2=FloatSlider(min=0.1, max=1, value=0.5),
                                  fase_2=FloatSlider(min=0, max=2*np.pi, value=np.pi))

# Mostrar gráfico interactivo
grafico_interactivo
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, FloatSlider

def plot_formas_de_onda(amplitud_1, frecuencia_1, fase_1, amplitud_2, frecuencia_2, fase_2):
    # Crear eje x
    x = np.linspace(0, 1, 1000)

    # Calcular formas de onda
    y_1 = amplitud_1 * np.sin(2 * np.pi * frecuencia_1 * x + fase_1)
    y_2 = amplitud_2 * np.sin(2 * np.pi * frecuencia_2 * x + fase_2)

    # Calcular la suma de las formas de onda
    y_suma = y_1 + y_2

    # Graficar formas de onda
    plt.plot(x, y_1, label='Onda 1')
    plt.plot(x, y_2, label='Onda 2')
    plt.plot(x, y_suma, label='Suma')

    # Agregar leyenda
    plt.legend()

# Crear controles deslizantes para la gráfica interactiva
grafica_interactiva = interactive(plot_formas_de_onda,
                                  amplitud_1=FloatSlider(min=0, max=1, value=0.5),
                                  frecuencia_1=FloatSlider(min=0.1, max=10, value=1),
                                  fase_1=FloatSlider(min=0, max=2*np.pi, value=0),
                                  amplitud_2=FloatSlider(min=0, max=1, value=0.5),
                                  frecuencia_2=FloatSlider(min=0.1, max=10, value=2),
                                  fase_2=FloatSlider(min=0, max=2*np.pi, value=np.pi))

# Mostrar la gráfica interactiva
grafica_interactiva

```
