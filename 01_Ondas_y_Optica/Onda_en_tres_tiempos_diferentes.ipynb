## 🌊 Onda en tres instantes diferentes

La imagen muestra una **onda sinusoidal** propagándose en el eje $x$ en tres momentos distintos:  
- $t = 0$  
- $t = T/2$  
- $t = T$  

Cada gráfico representa la función de onda:

\[
\psi(x,t) = y_m \sin(kx - \omega t)
\]

donde:
- $y_m$ es la amplitud,
- $k = \dfrac{2\pi}{\lambda}$ el número de onda,
- $\omega = \dfrac{2\pi}{T}$ la frecuencia angular.

### ¿Qué se observa?
- La **forma de la onda** se mantiene, pero **su posición cambia**: esto indica que la onda se **propaga hacia la derecha**.
- En cada instante se destaca en rojo **un ciclo completo** que se desplaza con el tiempo.
- Las flechas refuerzan la dirección de propagación.
- La franja roja muestra cómo **un mismo punto de la onda se mueve** a lo largo del eje $x$.

Este comportamiento ilustra la **propagación de una onda periódica**: la fase cambia con el tiempo, pero la amplitud y la longitud de onda permanecen constantes.


```python
import numpy as np
import matplotlib.pyplot as plt

# Parámetros de la onda
ym = 1        # amplitud
lam = 4       # longitud de onda
T = 1         # período
k = 2 * np.pi / lam
w = 2 * np.pi / T

# Rango espacial
x = np.linspace(0, 8, 500)

# Tiempos a graficar
tiempos = [0, T/2, T]
labels = [r'$t = 0$', r'$t = T/2$', r'$t = T$']

# Crear las gráficas
fig, axs = plt.subplots(3, 1, figsize=(8, 8), sharex=True)

for i, t in enumerate(tiempos):
    y = ym * np.sin(k * x - w * t)
    axs[i].plot(x, y, 'k', label='Onda completa')
    axs[i].set_ylim(-ym*1.2, ym*1.2)
    axs[i].set_ylabel(r'$\psi(x, t)$')
    axs[i].set_title(f'{labels[i]}')
    axs[i].grid(True)
    axs[i].axhline(0, color='black', linewidth=0.5)
    axs[i].axvline(0, color='black', linewidth=0.5)

    # Calcular el intervalo a remarcar (1 ciclo completo desplazado)
    x_start = (w * t) / k
    x_end = x_start + lam

    # Máscara del rango a remarcar
    mask = (x >= x_start) & (x <= x_end)
    axs[i].plot(x[mask], y[mask], color='red', linewidth=3, label='Sección destacada')

    axs[i].annotate('', xy=(7, 0.7), xytext=(6, 0.7),
                    arrowprops=dict(arrowstyle="->", lw=2))

axs[-1].set_xlabel('x')
plt.tight_layout()
plt.suptitle('Onda en tres tiempos diferentes', y=1.02, fontsize=16)
plt.show()
```
