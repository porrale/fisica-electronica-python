## 🌊 Ondas
### ↔️ ¿Qué es una onda longitudinal?

Una **onda longitudinal** es un tipo de onda en la cual las partículas del medio vibran en la **misma dirección** en la que se propaga la onda.

A diferencia de las ondas transversales (donde las partículas se mueven perpendicularmente al avance de la onda), en una onda longitudinal las zonas de **compresión** y **rarefacción** se alternan a lo largo del eje de propagación.

Un ejemplo clásico de onda longitudinal es el **sonido en el aire**, donde las moléculas de aire oscilan hacia adelante y hacia atrás a medida que la onda sonora avanza.

#### Características:
- Se propagan en sólidos, líquidos y gases.
- No tienen crestas ni valles, sino **zonas de compresión y rarefacción**.
- La energía se transmite en la misma dirección que las oscilaciones del medio.

> ⚠️ Las ondas longitudinales requieren un medio material para propagarse. No pueden viajar en el vacío.




```python
import matplotlib.pyplot as plt
import numpy as np

# Parámetros
num_lineas = 300
x_total = 12
x_base = np.linspace(0, x_total, num_lineas)
onda = np.sin(2 * np.pi * x_base / x_total * 5)  # 5 ciclos

# Variación senoidal de la separación (inversamente proporcional a la densidad)
separacion = 0.04 - 0.02 * onda  # menor separación en compresiones

# Calcular posición acumulada de cada línea
x_pos = np.cumsum(separacion)

# Crear figura
fig, ax = plt.subplots(figsize=(12, 2))

# Dibujar líneas verticales azules en posiciones moduladas
for x in x_pos:
    ax.plot([x, x], [0, 1], color='dodgerblue', linewidth=0.5)

# Flecha de amplitud
ax.annotate('Amplitud', xy=(x_pos[0], 1), xytext=(x_pos[0], 1.4),
            arrowprops=dict(arrowstyle='->', linestyle='dashed', color='dodgerblue'),
            ha='center', va='bottom', fontsize=12, color='dodgerblue')

# Etiqueta de onda
#ax.text(x_pos[-1], 1.1, 'Onda longitudinal', color='dodgerblue', ha='right', fontsize=9)

# Estética
ax.set_axis_off()
plt.title('Onda Longitudinal', fontsize=12, pad=20)
plt.tight_layout()
plt.show()
```

### ↕️ ¿Qué es una onda transversal?

Una **onda transversal** es aquella en la que las partículas del medio se mueven en dirección **perpendicular** a la propagación de la onda.

En este tipo de onda, se pueden identificar claramente los **puntos más altos (crestas)** y **más bajos (valles)**, así como la **línea de equilibrio**, que representa la posición sin perturbación.

#### Características:
- Se propagan en sólidos y en la superficie de líquidos.
- La **amplitud** es la distancia máxima desde la línea de equilibrio hasta una cresta o un valle.
- El ejemplo más común es una **onda en una cuerda**.

> 💡 A diferencia de las ondas longitudinales, las ondas transversales pueden propagarse sin comprimir el medio, desplazando energía de forma perpendicular a la oscilación.


```python
import matplotlib.pyplot as plt
import numpy as np

# Ejes
x = np.linspace(0, 4 * np.pi, 1000)
y = np.sin(x)

# Crear gráfico
fig, ax = plt.subplots(figsize=(10, 2))
ax.plot(x, y, color='royalblue', linewidth=2)

# Línea de equilibrio (eje horizontal punteado)
ax.axhline(0, color='gray', linestyle='dotted')

# Flecha de amplitud
ax.annotate('Amplitud', xy=(np.pi/2, 1), xytext=(np.pi/2, 1.3),
            arrowprops=dict(arrowstyle='->', color='purple'),
            ha='center', color='purple', fontsize=12)

# Texto de equilibrio
ax.text(0.5, 0.1, 'Equilibrio', color='purple', fontsize=12)

# Estética
ax.set_axis_off()
plt.title('Onda Transversal', fontsize=12, pad=20)
plt.tight_layout()
plt.show()
```
