## 🌊 Difracción e Interferencia en la Doble Rendija

Este experimento muestra cómo la luz (o cualquier onda) se comporta al pasar por **dos rendijas**:

- **Interferencia**: Las ondas provenientes de cada rendija se superponen, creando **franjas claras y oscuras** debido a la diferencia de fase.
- **Difracción**: Cada rendija actúa como una fuente que se dispersa, generando una **envolvente** que atenúa las franjas.

El patrón observado es la combinación de ambos efectos:
\[
I(\theta) \propto \cos^2\!\left(\frac{\pi d \sin\theta}{\lambda}\right) \cdot \text{sinc}^2\!\left(\frac{\pi a \sin\theta}{\lambda}\right)
\]
donde:
- $\lambda $: longitud de onda
- $ d $: separación entre rendijas
- $a $: ancho de cada rendija
- $\theta $: ángulo respecto al eje central

En la gráfica:
- **Curva sólida**: patrón total (interferencia + difracción)
- **Curva discontinua**: envolvente de difracción (rendija única)



```python
# Patrón de difracción e interferencia (doble rendija)
import numpy as np
import matplotlib.pyplot as plt

# Parámetros físicos
lambda_light = 500e-9  # longitud de onda (500 nm)
d = 2e-4              # separación entre rendijas (0.2 mm)
a = 1e-4              # ancho de cada rendija (0.1 mm)
L = 1.0               # distancia a la pantalla (1 m)

# Ángulo máximo y puntos
theta = np.linspace(-0.02, 0.02, 2000)  # radianes

# Factor de interferencia (dos rendijas)
beta = (np.pi * d * np.sin(theta)) / lambda_light
interference = (np.cos(beta))**2

# Factor de difracción (rendija única)
alpha = (np.pi * a * np.sin(theta)) / lambda_light
diffraction = (np.sinc(alpha/np.pi))**2

# Intensidad total
intensity = interference * diffraction

# Convertir ángulo a posición en la pantalla (x)
x = L * np.tan(theta) * 1000  # en mm

# Graficar
plt.figure(figsize=(10, 6))
plt.plot(x, intensity, label='Patrón total')
plt.plot(x, diffraction, '--', label='Envolvente (difracción)')
plt.title('Patrón de Difracción e Interferencia (Doble Rendija)')
plt.xlabel('Posición en la pantalla (mm)')
plt.ylabel('Intensidad relativa')
plt.legend()
plt.grid(True)
plt.show()
```
