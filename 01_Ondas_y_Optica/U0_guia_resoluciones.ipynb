## Problema 1
Dada la ecuación de una onda en una cuerda tensa:
\[ y(x, t) = 0.05 \sin(3x - 10t) \]
donde las magnitudes están en el sistema MKS. Determine:
1. La amplitud de la onda.
2. La velocidad de propagación.
3. La longitud de onda.
4. La frecuencia.
5. La dirección y sentido de propagación.

```python
# Resolución del Problema 1
import math

# Datos
A = 0.05  # Amplitud en metros
k = 3  # Número de onda en rad/m
omega = 10  # Frecuencia angular en rad/s

# Cálculos
v = omega / k  # Velocidad de propagación
lambda_ = 2 * math.pi / k  # Longitud de onda
f = omega / (2 * math.pi)  # Frecuencia

print(f"Amplitud: {A} m")
print(f"Velocidad de propagación: {v:.2f} m/s")
print(f"Longitud de onda: {lambda_:.2f} m")
print(f"Frecuencia: {f:.2f} Hz")
print("Dirección: eje x, sentido positivo")
```

## Problema 2
Considere la ecuación de una onda en una cuerda:
\[ y(x, t) = 0.03 \sin(3x - 2t) \]
Determine:
1. Para \( t=0 \), los valores de \( y \) para \( x = 0.1, 0.2, 0.3 \) m.
2. Para \( x=0.1 \) m, los valores de \( y \) para \( t=0, 0.1, 0.2 \) s.
3. La velocidad de oscilación de un tramo pequeño \( \Delta x \) de la cuerda.
4. La velocidad máxima en la oscilación.
5. La velocidad de propagación.

```python
# Resolución del Problema 2
A = 0.03  # Amplitud en metros
k = 3  # Número de onda en rad/m
omega = 2  # Frecuencia angular en rad/s

x_values = [0.1, 0.2, 0.3]
t_values = [0.0, 0.1, 0.2]

print("Valores de y para t=0 y x en [0.1, 0.2, 0.3]:")
for x in x_values:
    y = A * math.sin(k * x)
    print(f"x={x:.1f} m -> y={y:.4f} m")

print("\nValores de y para x=0.1 m y t en [0.0, 0.1, 0.2]:")
for t in t_values:
    y = A * math.sin(-omega * t)
    print(f"t={t:.1f} s -> y={y:.4f} m")

v_max = A * omega
v_propag = omega / k
print(f"\nVelocidad máxima de oscilación: {v_max:.4f} m/s")
print(f"Velocidad de propagación: {v_propag:.4f} m/s")
```

## Problema 3
Una cuerda de densidad lineal $ \mu = 0.3 $ kg/m está sometida a una tensión $ F = 60 $ N.
Si un extremo vibra con frecuencia $ f = 10 $ Hz y amplitud $ A = 1 $ cm, y la velocidad de propagación es:

$ v = \sqrt{\frac{F}{\mu}} $
Calcule:
1. La longitud de onda.
2. La ecuación de la onda armónica que se propaga hacia la derecha.

Vamos a resolver este problema paso a paso.

### Datos iniciales:

- Densidad lineal: \( \mu = 0.3 \, \text{kg/m} \)
- Tensión: \( F = 60 \, \text{N} \)
- Frecuencia: \( f = 10 \, \text{Hz} \)
- Amplitud: \( A = 1 \, \text{cm} = 0.01 \, \text{m} \)
- Velocidad de propagación: \( v = \sqrt{\frac{F}{\mu}} \)

---

### 1. Calcular la longitud de onda (\( \lambda \)):

La longitud de onda está relacionada con la velocidad de propagación (\( v \)) y la frecuencia (\( f \)) por la fórmula:
\[
\lambda = \frac{v}{f}
\]

Primero, calculamos la velocidad de propagación:
\[
v = \sqrt{\frac{F}{\mu}} = \sqrt{\frac{60}{0.3}} = \sqrt{200} \approx 14.14 \, \text{m/s}
\]

Luego, calculamos la longitud de onda:
\[
\lambda = \frac{v}{f} = \frac{14.14}{10} = 1.414 \, \text{m}
\]

---

### 2. Escribir la ecuación de la onda armónica:

La forma general de una onda armónica que se propaga hacia la derecha es:
\[
y(x, t) = A \sin(kx - \omega t)
\]

Donde:
- \( A = 0.01 \, \text{m} \) (amplitud),
- \( k = \frac{2\pi}{\lambda} \) (número de onda),
- \( \omega = 2\pi f \) (frecuencia angular).

Primero, calculamos el número de onda:
\[
k = \frac{2\pi}{\lambda} = \frac{2\pi}{1.414} \approx 4.44 \, \text{rad/m}
\]

Ahora, calculamos la frecuencia angular:
\[
\omega = 2\pi f = 2\pi \times 10 \approx 62.83 \, \text{rad/s}
\]

Finalmente, la ecuación de la onda es:
\[
y(x, t) = 0.01 \sin(4.44x - 62.83t)
\]

---

### Respuesta final:

1. La longitud de onda es:
\[
\lambda \approx 1.414 \, \text{m}
\]

2. La ecuación de la onda armónica es:
\[
y(x, t) = 0.01 \sin(4.44x - 62.83t)
\]

## Problema 4

Dos rendijas paralelas horizontales angostas (separadas una distancia a d= 6 nm) se iluminan con un haz de luz de 500 nm, como se muestra en la figura. La luz que se difracta a ciertos ángulos se refuerza; en otros, se cancela. Calcule los dos valores de más pequeños para los cuales haya interferencia constructiva.

![Descripción de la imagen](https://res.cloudinary.com/drruqi1kg/image/upload/v1733757305/Screenshot_5_rj2y8y.png)




## Fórmula clave:
La condición para la interferencia constructiva está dada por:

$$
d \sin\theta = m \lambda
$$

donde:
- $d$ es la separación entre las rendijas: $6 \, \text{nm} $,
- $\lambda$ es la longitud de onda de la luz: $\lambda = 500 \, \text{nm} = 5 \times 10^{-7} \, \text{m}$,
- $m$ es el orden de interferencia ($m = 1, 2, \dots$).

## Pasos de Resolución:
1. Calcular los ángulos $\theta$ para los dos valores más pequeños de $m$: $m = 1$ y $m = 2$.
2. Utilizar la fórmula para resolver:
   $$
   \theta = \arcsin\left(\frac{m \lambda}{d}\right)
   $$



```python
# Resolución del Problema 4
import numpy as np

# Datos del problema
d = 0.000006  # Separación entre rendijas en metros
lambda_luz = 500e-9  # Longitud de onda en metros

# Órdenes de interferencia
m_values = [0, 1, 2, 3]

# Calcular los ángulos de interferencia constructiva
theta = [np.arcsin(m * lambda_luz / d) for m in m_values]

# Convertir los ángulos a grados
theta_degrees = [np.degrees(theta) for theta in theta]

# Imprimir los resultados
for m, theta in zip(m_values, theta_degrees):
    print(f"Orden m={m}: ángulo θ = {theta:.4f} grados")
```

## Problema 5

Supongamos dos cargas puntuales

$(q_1 = 1 , \mu\text{C})$

y $(q_2 = -2 , \mu\text{C})$  

situadas a una distancia de $(r = 3 , \text{cm} = 0.03 , \text{m})$. Queremos calcular la magnitud de la fuerza entre ellas utilizando la Ley de Coulomb:

F = k \frac{q_1 q_2}{r^2} $

Donde:
- $(k = 8.987 \times 10^9 \, \text{N·m}^2/\text{C}^2)$ es la constante de Coulomb.
- $(q_1 = 1 \, \mu\text{C} = 1 \times 10^{-6} \, \text{C}$)
- $(q_2 = -2 \, \mu\text{C} = -2 \times 10^{-6} \, \text{C}$)
- $(r = 0.03 \, \text{m}$)

El signo del resultado nos indicará si la fuerza es atractiva o repulsiva.


```python
# Constantes y valores iniciales
k = 8.987e9  # Constante de Coulomb en N·m²/C²
q1 = 1e-6    # Carga 1 en Coulombs
q2 = -2e-6   # Carga 2 en Coulombs
r = 0.03     # Distancia en metros

# Cálculo de la fuerza
F = k * (q1 * q2) / r**2

# Resultado
print(f"La magnitud de la fuerza es: {F:.2f} N")
```

## Problema 5
Determine la intensidad de las ondas luminosas a una distancia de \( r = 1.5 \) m de una lámpara de \( P = 60 \) W, suponiendo que el 5% de la potencia se emite como luz visible y de manera uniforme en todas las direcciones.

```python
# Resolución del Problema 5
# Datos
P = 60  # Potencia total en W
eff = 0.05  # Eficiencia del 5%
r = 1.5  # Distancia en m

# Cálculos
P_visible = P * eff  # Potencia en luz visible
I = P_visible / (4 * math.pi * r**2)  # Intensidad

print(f"Intensidad de luz visible: {I:.4f} W/m^2")
```

## Problema 6
La distancia del Sol a Marte es aproximadamente \( \frac{3}{2} \) de la distancia del Sol a la Tierra. Sabiendo que la densidad de potencia que incide en las capas superiores de la atmósfera terrestre, denominada constante solar, es \( S_0 = 1366 \) W/m\(^2\), determine la densidad incidente en Marte.

```python
# Resolución del Problema 6
# Datos
S0 = 1366  # Constante solar en W/m^2
ratio = 3 / 2  # Relación de distancia Marte/Tierra

# Cálculos
SM = S0 / (ratio**2)  # Densidad de potencia en Marte

print(f"Densidad de potencia en Marte: {SM:.4f} W/m^2")
```
