## Problema 1
Dada la ecuación de una onda en una cuerda tensa:
\[ y(x, t) = 0.05 \sin(3x - 10t) \]
donde las magnitudes están en el sistema MKS. Determine:
1. La amplitud de la onda.
2. La velocidad de propagación.
3. La longitud de onda.
4. La frecuencia.
5. La dirección y sentido de propagación.

```python
# Resolución del Problema 1
import math

# Datos
A = 0.05  # Amplitud en metros
k = 3  # Número de onda en rad/m
omega = 10  # Frecuencia angular en rad/s

# Cálculos
v = omega / k  # Velocidad de propagación
lambda_ = 2 * math.pi / k  # Longitud de onda
f = omega / (2 * math.pi)  # Frecuencia

print(f"Amplitud: {A} m")
print(f"Velocidad de propagación: {v:.2f} m/s")
print(f"Longitud de onda: {lambda_:.2f} m")
print(f"Frecuencia: {f:.2f} Hz")
print("Dirección: eje x, sentido positivo")
```

## Problema 2
Considere la ecuación de una onda en una cuerda:
\[ y(x, t) = 0.03 \sin(3x - 2t) \]
Determine:
1. Para \( t=0 \), los valores de \( y \) para \( x = 0.1, 0.2, 0.3 \) m.
2. Para \( x=0.1 \) m, los valores de \( y \) para \( t=0, 0.1, 0.2 \) s.
3. La velocidad de oscilación de un tramo pequeño \( \Delta x \) de la cuerda.
4. La velocidad máxima en la oscilación.
5. La velocidad de propagación.

```python
# Resolución del Problema 2
A = 0.03  # Amplitud en metros
k = 3  # Número de onda en rad/m
omega = 2  # Frecuencia angular en rad/s

x_values = [0.1, 0.2, 0.3]
t_values = [0.0, 0.1, 0.2]

print("Valores de y para t=0 y x en [0.1, 0.2, 0.3]:")
for x in x_values:
    y = A * math.sin(k * x)
    print(f"x={x:.1f} m -> y={y:.4f} m")

print("\nValores de y para x=0.1 m y t en [0.0, 0.1, 0.2]:")
for t in t_values:
    y = A * math.sin(-omega * t)
    print(f"t={t:.1f} s -> y={y:.4f} m")

v_max = A * omega
v_propag = omega / k
print(f"\nVelocidad máxima de oscilación: {v_max:.4f} m/s")
print(f"Velocidad de propagación: {v_propag:.4f} m/s")
```

# Resolución del Problema 4

Suponga que en una cuerda de masa $m = 0.84 \, \mathrm{kg}$ y longitud $l = 3.8 \, \mathrm{m}$, se aplica una tensión de $T = 72 \, \mathrm{N}$.

1. Determine la frecuencia fundamental de las ondas estacionarias en la cuerda.
2. Calcule la frecuencia para producir una onda estacionaria con dos antinodos.

---

## Conceptos clave

### **Densidad lineal ($\mu$)**:
La densidad lineal de la cuerda ($\mu$) es la masa por unidad de longitud:
$
\mu = \frac{m}{l}
$
donde:
- $m$ es la masa de la cuerda en kilogramos (kg).
- $l$ es la longitud de la cuerda en metros (m).

La densidad lineal nos indica cuánta masa tiene la cuerda por cada metro de longitud y es fundamental para calcular la velocidad de propagación de las ondas.

---

### **Velocidad de propagación ($v$)**:
La velocidad de propagación de las ondas en la cuerda ($v$) depende de la tensión ($T$) aplicada y de la densidad lineal ($\mu$):
$
v = \sqrt{\frac{T}{\mu}}
$
donde:
- $T$ es la tensión en la cuerda (N).
- $\mu$ es la densidad lineal ($\mathrm{kg/m}$).

Esta ecuación proviene de la dinámica de ondas en un medio elástico, como una cuerda tensa.

---

### **Frecuencia fundamental ($f_1$)**:
La frecuencia fundamental ($f_1$) es la frecuencia más baja a la que la cuerda puede oscilar en un patrón de onda estacionaria. En este caso, la longitud de onda corresponde a dos veces la longitud de la cuerda:
$
f_1 = \frac{v}{2l}
$
donde:
- $v$ es la velocidad de propagación.
- $l$ es la longitud de la cuerda.

En este modo, la cuerda tiene un nodo en cada extremo y un único antinodo en el centro.

---

### **Frecuencia con dos antinodos ($f_2$)**:
Si la cuerda tiene dos antinodos, la frecuencia será el doble de la frecuencia fundamental:
$
f_2 = 2f_1
$
En este caso, la longitud de onda es igual a la longitud de la cuerda.

---

## Procedimiento para resolver el problema

1. **Calcular la densidad lineal ($\mu$)**:
   Usando:
 $
   \mu = \frac{m}{l}
   $

2. **Calcular la velocidad de propagación ($v$)**:
   Sustituye $\mu$ y $T$ en:
   $
   v = \sqrt{\frac{T}{\mu}}
 $

3. **Calcular la frecuencia fundamental ($f_1$)**:
   Utiliza:
  $
   f_1 = \frac{v}{2l}
 $
4. **Calcular la frecuencia con dos antinodos ($f_2$)**:
   Utiliza:
   $
   f_2 = 2f_1
   $

---

## Ejemplo numérico

- Masa de la cuerda: $m = 0.84 \, \mathrm{kg}$
- Longitud de la cuerda: $l = 3.8 \, \mathrm{m}$
- Tensión: $T = 72 \, \mathrm{N}$

1. **Densidad lineal**:
  $
   \mu = \frac{0.84}{3.8} \approx 0.221 \, \mathrm{kg/m}
  $

2. **Velocidad de propagación**:
   $
   v = \sqrt{\frac{72}{0.221}} \approx \sqrt{325.34} \approx 18.04 \, \mathrm{m/s}
  $

3. **Frecuencia fundamental**:
  $
   f_1 = \frac{18.04}{2 \times 3.8} \approx \frac{18.04}{7.6} \approx 2.37 \, \mathrm{Hz}
 $

4. **Frecuencia con dos antinodos**:
 $
   f_2 = 2 \times 2.37 \approx 4.75 \, \mathrm{Hz}
   $

---

## Conclusión

- Frecuencia fundamental: $f_1 \approx 2.37 \, \mathrm{Hz}$
- Frecuencia con dos antinodos: $f_2 \approx 4.75 \, \mathrm{Hz}$


```python
# Resolución del Problema 4
# Datos
m = 0.84  # Masa en kg
l = 3.8  # Longitud en m
T = 72  # Tensión en N

# Cálculos
mu = m / l  # Densidad lineal
v = math.sqrt(T / mu)  # Velocidad de propagación
f1 = v / (2 * l)  # Frecuencia fundamental
f2 = 2 * f1  # Frecuencia para 2 antinodos

print(f"Frecuencia fundamental: {f1:.4f} Hz")
print(f"Frecuencia con 2 antinodos: {f2:.4f} Hz")
```

## Problema 5
Determine la intensidad de las ondas luminosas a una distancia de \( r = 1.5 \) m de una lámpara de \( P = 60 \) W, suponiendo que el 5% de la potencia se emite como luz visible y de manera uniforme en todas las direcciones.

```python
# Resolución del Problema 5
# Datos
P = 60  # Potencia total en W
eff = 0.05  # Eficiencia del 5%
r = 1.5  # Distancia en m

# Cálculos
P_visible = P * eff  # Potencia en luz visible
I = P_visible / (4 * math.pi * r**2)  # Intensidad

print(f"Intensidad de luz visible: {I:.4f} W/m^2")
```

## Problema 6
La distancia del Sol a Marte es aproximadamente $ \frac{3}{2} $ de la distancia del Sol a la Tierra. Sabiendo que la densidad de potencia que incide en las capas superiores de la atmósfera terrestre, denominada constante solar, es $ S_0 = 1366 $ $W/m^2$, determine la densidad incidente en Marte.

```python
# Resolución del Problema 6
# Datos
S0 = 1366  # Constante solar en W/m^2
ratio = 3 / 2  # Relación de distancia Marte/Tierra

# Cálculos
SM = S0 / (ratio**2)  # Densidad de potencia en Marte

print(f"Densidad de potencia en Marte: {SM:.4f} W/m^2")
```
