```python
import numpy as np
import matplotlib.pyplot as plt

# Definir las constantes
c = 299792458  # velocidad de la luz en m/s
m = 1          # masa del objeto en kg

# Definir la función para calcular la energía relativista
def relativistic_energy(m, v):
    gamma = 1 / np.sqrt(1 - (v/c)**2)
    return gamma * m * c**2

# Generar un conjunto de velocidades
v = np.linspace(0, 0.99*c, 1000)

# Calcular la energía relativista para cada velocidad
energies = relativistic_energy(m, v)

# Graficar los resultados
fig, ax = plt.subplots()
ax.plot(v, energies)
ax.set_xlabel('Velocidad (m/s)')
ax.set_ylabel('Energía relativista (J)')
plt.grid()
plt.show()
```

Un cohete viaja a una velocidad constante de 0.6c (es decir, el 60% de la velocidad de la luz) con respecto a un observador en la Tierra. En el cohete, hay un reloj que marca el tiempo propio del cohete (es decir, el tiempo medido por un observador en el cohete). Si el cohete viaja durante 10 años según el tiempo propio del cohete, ¿cuántos años han pasado según el reloj de un observador en la Tierra?

Pista: Puedes usar la fórmula de dilatación del tiempo de la relatividad especial para resolver este problema.
La fórmula para el factor de Lorentz es:

$$\gamma = \frac{1}{\sqrt{1 - \frac{v^2}{c^2}}}$$

donde $v$ es la velocidad del cohete y $c$ es la velocidad de la luz.

La fórmula para la dilatación del tiempo es:

$$t_{tierra} = \gamma \cdot t_{cohete}$$

donde $t_{tierra}$ es el tiempo medido por un observador en la Tierra y $t_{cohete}$ es el tiempo propio del cohete.


```python
import numpy as np
from IPython.display import Latex

# Definir una fórmula en LaTeX
formula_latex = r'$\frac{1}{\sqrt{1 - \frac{v^2}{c^2}}}$'

# Mostrar la fórmula en la salida de la celda
display(Latex(formula_latex))

# Velocidad del cohete en unidades de c
v = 0.6

# Tiempo propio del cohete en años
tp = 10

# Factor de Lorentz
gamma = 1 / np.sqrt(1 - v**2)

# Tiempo medido por el observador en la Tierra
tt = tp * gamma

print("Según el observador en la Tierra, han pasado", round(tt, 3), "años.")
```

Supongamos que queremos calcular la energía necesaria para que un electrón pase del nivel n=1 al nivel n=3 en un átomo de hidrógeno. Sabiendo que la energía de un electrón en el nivel n está dada por la fórmula:

E = -13.6/n^2 eV

donde n es el número cuántico del nivel, y que la energía necesaria para excitar un electrón desde el nivel n1 al nivel n2 está dada por la fórmula:

∆E = E2 - E1 = (13.6/n1^2 - 13.6/n2^2) eV


#Resolución


```python
# E = -13.6/n^2 eV
# ∆E = E2 - E1 = (13.6/n1^2 - 13.6/n2^2) eV
```

```python
h = 6.626e-34 # Constante de Planck en J.s
c = 2.998e8 # Velocidad de la luz en m/s
e = 1.602e-19 # Carga del electrón en C
n1 = 3 # Nivel de energía inicial
n2 = 4 # Nivel de energía final
```

```python
delta_E = 13.6 * (1/n1**2 - 1/n2**2) # Energía necesaria en eV
delta_E_joules = delta_E * e # Energía necesaria en J
```

```python
# Calcular la frecuencia y la longitud de onda correspondientes
nu = delta_E_joules / h
lambda_ = c / nu
# Imprimir los resultados
print("La energía necesaria para excitar un electrón del nivel n=", n1, "al nivel n=", n2, "es de", delta_E_joules, "Joules.")
print("La frecuencia correspondiente es de", nu, "Hz, y la longitud de onda correspondiente es de", lambda_, "metros.")
```
