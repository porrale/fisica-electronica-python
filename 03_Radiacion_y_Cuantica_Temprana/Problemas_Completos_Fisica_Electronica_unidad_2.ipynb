# Problema 1: Emitancia Radiante del Cuerpo Negro
Calcular la emitancia radiante de un cuerpo negro a temperaturas $T = 400 \, \text{K}$ y $T = 3500 \, \text{K}$. Utilizar la Ley de Stefan-Boltzmann.

```python
from scipy.constants import sigma
def emitancia_radiante(T):
    return sigma * T**4

# Temperaturas
T1 = 400  # Kelvin
T2 = 3500  # Kelvin

# Emitancia
E1 = emitancia_radiante(T1)
E2 = emitancia_radiante(T2)
E1, E2
```

# Problema 2: Máxima Emisión
Determinar la longitud de onda para la máxima emisión de un cuerpo negro a temperaturas $T = 500\, \text{K}$ y $T = 5000 \, \text{K}$. Utilizar la Ley de Wien.

# Problema 4: Ley de Planck
Verificar que la fórmula de Planck se aproxima a la de Rayleigh-Jeans para longitudes de onda largas ($\lambda \to \infty$).

```python
import numpy as np
from scipy.constants import h, c, k

def planck_lambda(long_wavelength, T):
    return (2 * h * c**2) / (long_wavelength**5 * (np.exp(h * c / (long_wavelength * k * T)) - 1))

def rayleigh_jeans_lambda(long_wavelength, T):
    return (2 * c * k * T) / long_wavelength**4

# Comparar ambas fórmulas
lambda_long = 0.01  # Longitud de onda larga en metros
T = 300  # Kelvin
planck = planck_lambda(lambda_long, T)
rayleigh = rayleigh_jeans_lambda(lambda_long, T)
planck, rayleigh
```

# Problema 6: Experimento Fotoeléctrico
Determinar la energía cinética máxima de los electrones emitidos si la longitud de onda incidente es $300 \, \text{nm}$ sobre un cátodo de cesio con función trabajo $W = 2.1 \, \text{eV}$.

```python
def energia_cinetica(lambda_incidente, W):
    lambda_m = lambda_incidente * 1e-9  # Convertir a metros
    energia_foton = h * c / lambda_m
    energia_cinetica = energia_foton - W * 1.60218e-19  # Convertir W de eV a J
    return energia_cinetica / 1.60218e-19  # Convertir de J a eV

# Parámetros
lambda_incidente = 300  # nm
W = 2.1  # eV

E_k = energia_cinetica(lambda_incidente, W)
E_k
```

# Problema 8: Densidad de Potencia
Un láser de He-Ne emite a $\lambda = 632.82 \, \text{nm}$ con potencia $5 \, \text{mW}$. Determinar su densidad de potencia si el diámetro del rayo es $2 \, \text{mm}$.

```python
def densidad_potencia(P, d):
    area = 3.14159 * (d / 2)**2  # Área del rayo (m²)
    return P / area

# Parámetros
P = 5e-3  # Potencia en W
d = 2e-3  # Diámetro en m

densidad = densidad_potencia(P, d)
densidad
```
