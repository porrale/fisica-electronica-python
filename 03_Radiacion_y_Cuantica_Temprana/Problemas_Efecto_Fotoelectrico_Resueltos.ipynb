## Ejercicio 1: Energía cinética del fotoelectrón
Una luz de frecuencia $f = 7.5 \times 10^{14}\ \text{Hz}$ incide sobre una superficie de potasio, cuyo trabajo de extracción es $W = 2{,}29\ \text{eV}$. Calcular la energía cinética máxima de los electrones.

**Datos útiles:**
- $h = 6{,}626 \times 10^{-34}\ \text{J·s}$
- $1 \ \text{eV} = 1{,}602 \times 10^{-19}\ \text{J}$

### **Resolución**
$$
E_f = h f = W + E_c
$$
$$
E_c = e \cdot U_0
$$


```python
# @title
h = 6.626e-34  # J·s
e = 1.602e-19  # C
f = 7.5e14     # Hz
W_eV = 2.29
W_Joule = W_eV * e

Ef = h * f
Ec = Ef - W_Joule


print(f"Energía del fotón: {Ef:.2e} J")

print(f"Energía Cinetica max. electrones: {Ec:.2e} J")
```

## Ejercicio 2: Potencial de corte con luz violeta
Una luz de longitud de onda $\lambda = 436\ \text{nm}$ incide sobre una superficie de potasio. Calcular el potencial de frenado necesario para detener los fotoelectrones emitidos.

**Datos útiles:**
- $c = 3{,}00 \times 10^8 \ \text{m/s}$
- $h = 6{,}626 \times 10^{-34} \ \text{J·s}$
- $e = 1{,}602 \times 10^{-19} \ \text{C}$
- $W_c = 2{,}29 \ \text{eV}$

###Resolución
$f = \frac{c}{\lambda}$

$E_f = h \cdot f$

$E_c = E_f - W_c$

$U_0 = \frac{E_c}{e}$






```python
h = 6.626e-34  # J·s
e = 1.602e-19  # C
lambda_nm = 436  # nm
lambda_m = lambda_nm * 1e-9
c = 3e8  # m/s

f = c / lambda_m
Ef = h * f
Wc = 2.29 * e

Ec = Ef - Wc
U0 = Ec / e

print(f"Frecuencia: {f:.2e} Hz")
print(f"Energía del fotón: {Ef:.2e} J")
print(f"Potencial de corte U0: {U0:.2f} V")
```

## Ejercicio 3: Calcular frecuencia umbral y verificar ecuación de Einstein
Sabemos que la energía cinética máxima de los electrones emitidos por una luz de frecuencia $f = 6{,}02 \times 10^{14} \, \text{Hz}$ es $E_c = 0{,}2 \, \text{eV}$. El material utilizado es potasio, cuyo trabajo de extracción es $W_c = 2{,}29 \, \text{eV}$.

1. Verificá que se cumple la ecuación de Einstein:
$E_f = E_c + W_c$  (expresado en $[eV]$)
2. Calculá la frecuencia umbral $f_0$ necesaria para que comience a producirse el efecto fotoeléctrico.

**Datos útiles:**
- $h = 6{,}626 \times 10^{-34}\ \text{J·s}$
- $1\ \text{eV} = 1{,}602 \times 10^{-19}\ \text{J}$

###Resolución
Ef_eV = Ec_eV + Wc_eV

Wc_J = Wc_eV * e

f0 = Wc_J / h


```python
h = 6.626e-34  # J·s
e = 1.602e-19  # C
# 1eV == 1.602e-19  J
f = 6.0e14  # Hz
Ec_eV = 0.2
Wc_eV = 2.29

Ef_eV = Ec_eV + Wc_eV

print(f"Energía del fotón: {Ef_eV:.2f} eV")

# Frecuencia umbral
Wc_J = Wc_eV * e
frecuencia = (Ef_eV  * e )/h
f0 = Wc_J / h
print(f"Frecuencia umbral f0: {f0:.2e} Hz")
print(f"frecuencia foton total : {frecuencia:.2e} Hz")
```
