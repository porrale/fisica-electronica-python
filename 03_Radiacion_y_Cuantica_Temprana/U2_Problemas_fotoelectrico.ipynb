#Problema Fotoeléctrico
Una superficie de sodio se ilumina con luz que tiene una longitud de onda de 300 nm. La función de trabajo para el metal
sodio es 2.46 eV.
Calculas energía cinetica máxima y longitud de onda de corte.


#solución
Imagine un fotón que golpea la superficie metálica y expulsa un electrón. El electrón con la máxima energía
es uno cerca de la superficie que no experimenta interacciones con otras partículas en el metal que reduciría su energía en su
ruta fuera del metal.


Encuentre la energía de cada fotón en el haz de luz iluminante

 <font size="5"> $E=h.f=\frac{hc}{\lambda}$ </font>

A partir de la ecuación  halle la energía cinética máxima de un electrón:

 <font size="5"> $Ec_{max}=\frac{hc}{\lambda}-\phi$

 Calcular la longitud de onda de corte

 $\lambda_c = \frac{hc}{\phi}$

```python
import math

# Constantes
longitud_onda = 300 * 10**-9  # Longitud de onda en metros
funcion_trabajo = 2.46  # Función de trabajo en eV
carga_electron = 1.602176634 * 10**-19  # Carga del electrón en Coulombs
velocidad_luz = 2.99792458 * 10**8  # Velocidad de la luz en metros/segundo
constante_planck = 6.62607015 * 10**-34  # Constante de Planck en J·s

# Cálculo de la energía del fotón incidente
energia_foton = (velocidad_luz * constante_planck) / longitud_onda

# Conversión a electronvoltios (eV)
energia_foton_eV = energia_foton / carga_electron

# Cálculo de la energía cinética máxima
energia_cinetica_maxima = energia_foton_eV - funcion_trabajo
# Cálculo de la longitud_onda_corte
longitud_onda_corte = constante_planck * velocidad_luz / (funcion_trabajo * carga_electron)


# Imprimir resultado
print("La energía cinética máxima de los electrones es:", energia_cinetica_maxima, "eV")
print("la longitud_onda_corte:", longitud_onda_corte * 1e9, "nm")
```
