```python
#!/usr/bin/env python
# -*- coding: utf-8 -*-
from matplotlib import pyplot as plt
# Función lineal
def f1(x):
    return (0.5*x-1)


# Valores del eje X que toma el gráfico.
x = range(0, 40)
# Graficar  funcion.
plt.plot(x, [f1(i) for i in x],'o-')
# Establecer el color de los ejes.
plt.axhline(0, color="black")
plt.axvline(0, color="black")

# Limitar los valores de los ejes.
plt.xlim(0, 40)
plt.ylim(0, 24)

plt.xlabel('Z numero Atómico')        # Etiqueta del eje OX
plt.ylabel('Raiz de frecuencia')        # Etiqueta del eje OY
plt.title(' Ley de Moseley')    # Título del gráfico
print(x)
# Guardar gráfico como imágen PNG.
plt.savefig("output.png")
# Mostrarlo.
plt.grid()
plt.show()
```
