```python
def bremsstrahlung(i, u, z):
    """
  Intensidad de frenado (bremsstrahlung.)

    Args:
        i (float): Corriente [A]
        u (float):  Tension  [U]
        z (int): Numero Z

    Returns:
        float: Intensidad
    """
    c = 2.998 # velocidad de la luz 2,998e+8
    lambda_0 = 12390/U #Longitud de onda crítica [A] Ángstrom
    print(str(lambda_0)+" Ángstrom",U)
    lambdas = np.arange(lambda_0, 3,0.001)
    return lambdas, c*c/lambda_0*i*z*(lambdas-lambda_0)/lambdas**3
```

${\displaystyle {\frac {\mathrm {d} E}{\mathrm {d} \lambda }}=KIZ{\frac {h}{c}}\left({\frac {\lambda }{\lambda _{\mathrm {min} }}}-1\right){\frac {1}{\lambda ^{3}}}}$

```python
import numpy as np
import matplotlib.pyplot as plt
from pprint import pprint

def bremsstrahlung(i, u, z):
    c = 2.998  # velocidad de la luz [m/s]
    lambda_0 = 12390 / u  # Longitud de onda crítica [A] Ángstrom
    print(str(lambda_0) + " Ángstrom", u)
    lambdas = np.arange(lambda_0, 3, 0.001)
    return lambdas, c * c / lambda_0 * i * z * (lambdas - lambda_0) / lambdas**3

I = 0.0027  # Corriente [A]
elementos = {
    'W': {'z': 78}
}

plt.figure(figsize=(10, 10))
plt.title("Intensidad Bremsstrahlung Frenado Wolframio o tungsteno")

tensiones = [25e3, 35e3, 50e3]

for tension in tensiones:
    for elemento_nombre, elemento_dato in elementos.items():
        lambdas, intensidad = bremsstrahlung(I, tension, elemento_dato['z'])
        plt.plot(lambdas, intensidad, label=elemento_nombre + " " + str(tension) + " V", linewidth=2)

plt.minorticks_on()
plt.grid(which='major', color='#666666', linestyle='-')
plt.grid(which='minor', color='#999999', linestyle='-')
plt.legend()
plt.xlabel('$\lambda$, A')
plt.ylabel('Intensidad relativa')
# Hacer más intensos los ejes X e Y
plt.axhline(0, color='black', linewidth=2)  # Eje X
plt.axvline(0, color='black', linewidth=2)  # Eje Y
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def bremsstrahlung(i, u, z):
    c = 2.998  # velocidad de la luz [m/s]
    lambda_0 = 12390 / u  # Longitud de onda crítica [A] Ángstrom
    lambdas = np.arange(lambda_0, 3, 0.001)
    frenado_intensity = c**2 / lambda_0 * i * z * (lambdas - lambda_0) / lambdas**3

    characteristic_intensity_1 = np.exp(-(lambdas - .72) ** 2 / 0.001) * 20
    characteristic_intensity_2 = np.exp(-(lambdas - 1.) ** 2 / 0.001) * 20

    total_intensity = frenado_intensity + characteristic_intensity_1 + characteristic_intensity_2

    return lambdas, total_intensity

I = 0.0027  # Corriente [A]
elementos = {
    'W': {'z': 78}
}

plt.figure(figsize=(10, 10))
plt.title("Intensidad  Frenado  + caracteristico ")

tension = 50e3

for elemento_nombre, elemento_dato in elementos.items():
    lambdas, intensidad = bremsstrahlung(I, tension, elemento_dato['z'])
    plt.plot(lambdas, intensidad, label=elemento_nombre + " " + str(tension) + " V")

plt.minorticks_on()
plt.grid(which='major', color='#666666', linestyle='-')
plt.grid(which='minor', color='#999999', linestyle='-')
plt.legend()
plt.xlabel('$\lambda$, A')
plt.ylabel('Intensidad relativa')
plt.show()
```
