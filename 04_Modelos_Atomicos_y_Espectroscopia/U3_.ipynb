Niveles de energía Átomo cuentas auxiliares

Este código utiliza la fórmula ΔE = hν (donde ΔE es la diferencia de energía entre dos niveles de energía, h es la constante de Planck y ν es la frecuencia) para calcular las longitudes de onda de las líneas de emisión de la serie de Balmer. Luego, convierte la frecuencia en longitud de onda utilizando la velocidad de la luz.
Series de Balmer

$\Delta E =h.f = \left (\frac{1}{p{^2}} -\frac{1}{q{^2}}\right )13.6eV$


```python
#@title resolucion
import numpy as np

def serie_balmer(n):
    E_0 = 13.6  # Energía de ionización del hidrógeno en eV
    h = 4.135667696e-15  # Constante de Planck en eV*s

    # Calcular las longitudes de onda utilizando la fórmula de Balmer
    longitudes_onda = []
    for m in range(n + 1, n + 16):
        delta_E = E_0 * (1/(n**2) - 1/(m**2))
        print(delta_E)
        frecuencia = delta_E / h
        longitud_onda = 3e8 / frecuencia  # Convertir frecuencia a longitud de onda (c = velocidad de la luz en m/s)
        longitudes_onda.append(longitud_onda)

    return longitudes_onda

# Calcular las longitudes de onda para la serie de Balmer con n = 2
n = 2
longitudes_onda_balmer = serie_balmer(n)

# Convertir las longitudes de onda a nanómetros
longitudes_onda_balmer_nm = [longitud_onda * 1e9 for longitud_onda in longitudes_onda_balmer]

# Imprimir las longitudes de onda calculadas en nanómetros
for idx, longitud_onda_nm in enumerate(longitudes_onda_balmer_nm):
    print(f"Línea {idx + 3}: {longitud_onda_nm:.4f} nm")


```

# Radio  fundamental Hidrogeno

Enunciado:

Se desea calcular el radio del átomo de hidrógeno en su estado fundamental. Utilizando la fórmula del modelo atómico de Bohr, se puede obtener el radio del átomo de hidrógeno en el estado fundamental (r_1) a partir de las constantes físicas.

La fórmula para calcular el radio del átomo de hidrógeno en el estado fundamental es:

$
r_1 = \frac{{\epsilon_0 \cdot h^2}}{{\pi \cdot m \cdot e^2}}
$

donde:
- r_1 es el radio del átomo de hidrógeno en el estado fundamental.
- \epsilon_0 es la permitividad eléctrica del vacío.
- h es la constante de Planck.
- m es la masa del electrón.
- e es la carga del electrón.

Las constantes físicas a utilizar son:
- Permitividad eléctrica del vacío (\epsilon_0) = 8.85418782 \times 10^{-12} F/m.
- Constante de Planck (h) = 6.62607015 \times 10^{-34} J \cdot s.
- Masa del electrón (m) = 9.10938356 \times 10^{-31} kg.
- Carga del electrón (e) = 1.602176634 \times 10^{-19} C.

Calcula el radio del átomo de hidrógeno en su estado fundamental.



```python
import numpy as np

def calcular_radio_bohr():
    epsilon_0 = 8.85418782e-12  # Permitividad eléctrica del vacío en F/m
    h = 6.62607015e-34  # Constante de Planck en J*s
    m = 9.10938356e-31  # Masa del electrón en kg
    e = 1.602176634e-19  # Carga del electrón en C

    # Calcular el radio del átomo de hidrógeno en el estado fundamental (n = 1)
    r_1 = (epsilon_0 * h**2) / (np.pi * m * e**2)

    return r_1

# Calcular el radio del átomo de hidrógeno en el estado fundamental (n = 1)
radio_bohr = calcular_radio_bohr()

# Imprimir el resultado
print(f"El radio del átomo de hidrógeno en el estado fundamental (n = 1) es: {radio_bohr:.4e} m")


```

# Radios del hidrógeno
Se desea calcular los radios del átomo de hidrógeno en los estados excitados para los valores de n desde 2 hasta 7. Utilizando la fórmula del modelo atómico de Bohr, se puede obtener el radio del átomo de hidrógeno en el estado n (r_n) a partir del radio del estado fundamental (r_1) y el número atómico (Z).

La fórmula para calcular el radio del átomo de hidrógeno en el estado n es:


$r_n = r_1 \cdot \left(\frac{{n^2}}{{Z}}\right)$
# New Section

donde:
- r_n es el radio del átomo de hidrógeno en el estado n.
- r_1 es el radio del átomo de hidrógeno en el estado fundamental.
- n es el número cuántico principal del estado excitado.
- Z es el número atómico del hidrógeno.

Para realizar los cálculos, se utilizarán las siguientes constantes físicas:
- Permitividad eléctrica del vacío (\epsilon_0) = 8.85418782 \times 10^{-12} F/m.
- Constante de Planck (h) = 6.62607015 \times 10^{-34} J \cdot s.
- Masa del electrón (m) = 9.10938356 \times 10^{-31} kg.
- Carga del electrón (e) = 1.602176634 \times 10^{-19} C.

Calcula los radios del átomo de hidrógeno en los estados excitados para n = 2, 3, 4, 5, 6, y 7.



```python
import numpy as np

def calcular_radios_hidrogeno(n_min, n_max, Z):
    epsilon_0 = 8.85418782e-12  # Permitividad eléctrica del vacío en F/m
    h = 6.62607015e-34  # Constante de Planck en J*s
    m = 9.10938356e-31  # Masa del electrón en kg
    e = 1.602176634e-19  # Carga del electrón en C

    # Calcular el radio del átomo de hidrógeno en el estado fundamental (n = 1)
    r_1 = (epsilon_0 * h**2) / (np.pi * m * e**2)

    # Calcular los radios para n desde n_min hasta n_max
    radios_hidrogeno = []
    for n in range(n_min, n_max + 1):
        radio = r_1 * (n**2 / Z)
        radios_hidrogeno.append(radio)

    return radios_hidrogeno

# Datos del problema
n_min = 2  # Valor mínimo de n
n_max = 12  # Valor máximo de n
Z = 1  # Número atómico del hidrógeno

# Calcular los radios del hidrógeno para n desde 2 a 7
radios_hidrogeno = calcular_radios_hidrogeno(n_min, n_max, Z)

# Imprimir los resultados
for i, radio in enumerate(radios_hidrogeno):
    n = i + n_min
    print(f"Radio para n = {n}: {radio:.4e} m")
```
