```python
import matplotlib.pyplot as plt  # Paquete gráfico, se grafican las funciones a través de arrays.
from sympy import *              # Importa todas las funciones de sympy
init_printing()                  # Edita las fórmulas
x=symbols('x')                   # Símbolos con los cuales se trabajará algebraicamente
import numpy as np

# Función $\psi_n(x)=\sum_{k=1}^n(1/n)cos(k\pi x)$, donde $n$ índica el número de ondas planas que se superponen para construirla.

def Sup(z,n):
  s=0
  for k in range(1,n+1):
    s=s+(1/n)*np.cos(k*np.pi*z)
  return(s)

# Superp(n) realiza la gráfica de la función de onda $\psi_n(x)=\sum_{k=1}^n(1/n)cos(k\pi x)$, donde $n$ índica el número de ondas planas que se superponen para construirla.

def Superp(n):
  z=np.linspace(-1,1,100)
  fig, ax = plt.subplots()
  ax.plot(z,Sup(z,n),label='$\psi(x)$')
  ax.set_ylabel('Amplitud')
  ax.set_title('Función de onda')
  ax.legend()

# Superpk(n) reproduce la secuencia de gráficas obtenidas con Superp(n) desde 1 hasta n-1

def Superpk(n):
    for k in range(1,n+1):
        Superp(k)
```

# **NATURALEZA DUAL DE LA MATERIA**

En 1924 el físico Louis De Broglie en su convencimiento de que la naturaleza es simétrica realizo el siguiente razonamiento:
$$\text{'La luz tiene naturaleza dual, en algunos casos se comporta como ondas y en otros como partículas. Si la naturaleza es simétrica, esa dualidad también debería ser válida para la materia'}$$
Si una partícula actúa como una onda, debería tener una longitud de onda y una frecuencia. De está forma postulo que una partícula libre con masa en reposo $m$, que se mueve a una rapidez $v$ no relativista, debería tener una longitud de onda $\lambda$ relacionada con la cantidad de movimiento $p=mv$ exactamente de la misma forma que un fotón, como expresa la ecuación $\lambda=\frac{h}{p}$. Entoces la longitud de onda de De Broglie para una pratícula es $\lambda=\frac{h}{p}=\frac{h}{mv}$, donde $h$ es la constante de Planck. Según De Broglie, la frecuencia $\nu$ también se relaciona con la energía de la partícula $E$ en la misma forma que para un fotón, es decir, $E=h\nu$.

# **PRINCIPIO DE INCERTIDUMBRE DE HEISENBERG**

La hipótesis ondulatoria de De Broglie recibió confirmación experimental en 1927 por Clinton Davisson y Lester Germer. El descubrimiento de la naturaleza dual de la materia forzo a los físicos a reevaluar el lenguaje cinemático usado para describir la posición y el momento lineal de una partícula. De acuerdo con la mecánica clásica, no hay un límite que impida refinar los aparatos de medida o los procedimientos experimentales para alcanzar cualquier precisión deseada en la medición de los observables, como por ejemplo la posción y momento lineal, de una partícula. La teoría cuántica por el contrario, predice que es imposible realizar mediciones simultáneas de la posición ($x$) y momento lineal ($p_x$) de una partícula, de tal forma que la incertidumbre $\Delta x$ y $\Delta p_x$, respectivamente, de las mediciones sean arbitrariamente pequeñas. Este principio lo propuso Werner Heisenberg en 1927, haciendo notar que las incertidumbre inevitables en $\Delta x$ y $\Delta p_x$ no se deben a las imperfecciones en los instrumentos de medida, sino a la naturaleza cuántica de la materia. Matemáticamente el $\textbf{principio de incertidumbre de Heisenberg}$ se expresa:

$$\Delta x \Delta p_x\geq\frac{\hbar}{2}$$

es decir, si se realiza una medida de la posición de una partícula con incertidumbre $\Delta x$ y una medida simultánea de su momento lineal con incertidumbre $\Delta p_x$, ambas incertidumbres no
pueden ser más pequeñas que $\hbar/2$.

Para comprender la naturaleza del problema consideremos un haz de electrones en el vacío, con longitud de onda $\lambda$, disparados hacia ua rendija de ancho $a$ con $a\gg \lambda$. Como resultado de la naturaleza ondulatoria de la materia se obtiene el patrón de difracción que se ilustra en la siguiente figura. Aproximadamente el $85$% de los electrones llegan a la película dentro del máximo central y el resto llegan a ella dentro de los máximos subsidiarios en ambos lados. En este experimento no se deben considerar los electrones como partículas, sino como ondas, puesto que no todos siguen la misma trayectoría, aunque todos ellos tengan el mismo estado inicial de movimiento. Más aún, no podemos predecir la trayectoría exacta de un electrón aunque conozcamos su estado inicial, solamente podemos saber la probabilidad que un determinado electrón llegue a diversas regiones de la película.

Un electrón que llega a la película en la orilla externa del máximo central, formando el ángulo $\theta_1$, debe tener una componente de cantidad de movimiento $p_y$ en la dirección vertical $p_y$, y también una componente $p_x$ en la dirección $p_x$, aunque el haz de electrones se haya dirigido paralelo al eje $x$. Como podemos observar en  la figura estas dos componentes se relacionan por la expresión $tan\theta_1=\frac{p_y}{p_x}$. Como $\theta_1$ es pequeño en comparación con el ancho de la rendija $tan\theta_1\approx\theta_1$ y $p_y=p_x\theta_1$ y puesto que un patrón de interferencia presenta una franja oscura siempre que $sen\theta_1=\pm \frac{\lambda}{a}$, tenemos que $\theta_1=\frac{\lambda}{a}$ (puesto que $sen\theta_1\approx\theta_1$). Por lo tanto

$$p_y=\frac{p_x\lambda}{a}$$

Para el $85$% de los electrones que llegan a la película dentro del máximo central (entre los ángulos $-\lambda/a y \lambda/a$), la componente $y$ de la cantidad de movimiento se repartirá dentro de los límites $-p_x\lambda/a$ y $p_x\lambda/a$. Si pensamos en $\textbf{todos}$ los electrones que pasan por la rendija $(p_y)_{prom}=0$ a causa de la simetría del patrón de difracción. Por lo tanto la incertidumbre en la componente $y$ de la cantidad de movimiento será $\Delta p_y$ será por lo menos $p_x\lambda/a$, es decir

$$\Delta p_y\geq \frac{p_x\lambda}{a}$$

Note que cuanto menor sea el ancho de la rendija (es decir, hay menor incertidumbre en la componente $y$ de la posición de los electrones), más ancha será la figura de la difracción y mayor será la incertidumbre de la componente $y$ de la cantidad de movimiento $p_y$. Por la relación de De Broglie $\lambda=\frac{h}{p_x}$, tenemos que $\frac{p_x h}{ap_x}=\frac{h}{a}$, es decir

$$\Delta p_y\geq \frac{h}{a}\Rightarrow \Delta p_ya\geq h\Rightarrow \Delta p_y\Delta y\geq h \ (a=\Delta y)$$

En conclusión: el ancho $a$ de la rendija representa la incertidumbre en la componente $y$ de la posición de un electrón, al pasar por la rendija, puesto que no conocemos con exactitud dónde atraviesa cada partícula la rendija. En este caso hay incertidumbre tanto en la componente $y$ de la posición como en la componente $y$ de la cantidad de movimiento $p_y$ y las dos incertidumbres
se relacionan por la expresión $\Delta p_y a\geq h$, donde $a=\Delta y$. Se puede reducir la incertidumbre $\Delta p_y$ en la cantidad de movimiento sólo reduciendo el ancho del patrón de difracción para lo cual hay que aumentar el ancho $a$ de la rendija, lo cual aumenta la incertidumbre en la posición. Reciprocamente, cuando se disminuye la incertidumbre en la posición estrechando la rendija, el patrón de difracción se ensancha y aumenta la correspondiente incertidumbre en la cantidad de movimiento.

$\textbf{Nota}:$ Observe que la expresión $\Delta p_y\Delta y\geq h$ no es contradictoria con la formulación original $\Delta p_y\Delta y\geq \hbar/2$ del principio de incertidumbre, puesto que $h\geq\hbar/2$, siendo la $\hbar/2$ la cota mínima para el producto de las incertidumbres.

## **PARTÍCULA LIBRE**

El movimiento de una partícula libre, con masa $m$ y energía total $E$, en una región unidimensional a potencial constante $V$, puede ser descrito por medio de la función de onda $\psi(x)=Ae^{ikx}$, donde $A$ es la constante de normalización y el número de onda $k$ es igual a: $\frac{\sqrt{2m(E-V)}}{\hbar}$. Si utilizamos la identidad de Euler $e^{i\theta}=cos\theta+isen\theta$, la función de onda $\psi(x)=Ae^{ikx}$ se puede expresar de la forma $\psi(x)=A(coskx+isenkx)$. Tanto el coseno como el seno son ondas con longitud de onda $\lambda=\frac{2\pi}{k}$, por lo tanto $\frac{2\pi}{k}=\frac{h}{p}$ (usando la relación de De Broglie). Si despejamos la cantidad de movimiento $p$, tenemos que $p=k\hbar$, lo cual muestra que el momento lineal de la partícula está bien definido.

Aunque es posible conocer el momento lineal de la partícula su posición es totalmente impredecible, para lo cual es suficiente considerar la función de densidad de probabilidad $|\psi(x)|^2=A^2((coskx)^2+(senkx)^2)=A^2$, la cual es constante y por lo tanto la probabilidad de encontrar la partícula en cualquier porción de la recta es igual para segmentos de la misma longitud.

Si una partícula tiene una posición definida su función de onda debería tener el siguiente aspecto:

```python
Superp(20)
```

Como se observa en la figura, la partícula estaría localizada en la región donde la función de onda presenta un pico, puesto que allí se concentra la densida de probabilidad para su posición. estas funciones son el resultado de la superposición de un número grande de ondas planas $e^{ikx}$. Concluyendo, la función de onda $\psi(x)$ de una partícula bien localizada se puede representar como una sumatoria (superposición) de ondas planas con diferente momento lineal, es decir, $\psi(x)=\sum_{j=1}^nA_je^{ik_jx}$. En este caso, aunque la certidumbre en la localización aumenta, la incertidumbre en el momento lineal se indeterminará a medida que se superpongan cada vez más ondas planas con diferentes momentos lineales.
