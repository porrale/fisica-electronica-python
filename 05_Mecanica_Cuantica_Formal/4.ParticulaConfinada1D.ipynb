# **PARTÍCULA CONFINADA EN UNA CAJA UNO DIMENSIONAL**

Por partícula en una caja 1-dimensional se entiende una partícula de masa $m_p$ que se mueve en una dimensión, la cual está sujeta a una función de energía potencial que tiende a infinito en todas partes excepto en una región de longitud $l$, en donde toma un valor de cero. La siguiente gráfica porporciona una imagen visual de un pozo de potencial con amplitud $l=2$.

```python
#@title
plt.plot([-0.3,0,0,l,l,l+0.3],[V_0,V_0,0,0,V_0,V_0],color='blue')
plt.xlabel('Amplitud de la caja ($a_0$) ')
plt.ylabel('Energía (Ha)')
plt.title('Pozo de potencial 1-dimensional')
plt.fill_between([-0.3,0],[V_0,V_0],color='blue')
plt.fill_between([l,l+0.3],[V_0,V_0],color='blue')
```

El hamiltoniano para este sistema de una partícula con masa $m_p$, está dado por $H=-\frac{\hbar²}{2m_p}\frac{d²}{dx²}+V(x)$, donde $V(x)=0$, para $0\leq x \leq l$ y $V(x)=\infty$ en otra parte. Si $\psi$ es la función de onda para está partícula, las soluciones a la ecuación de $Schr\ddot{o}dinger$:

$$
H\psi=E\psi
$$

están dada por $\psi_n(x)=\sqrt{\frac{2}{l}}sin(\frac{n\pi x}{l})$, en donde $n$ toma cualquier valor entero mayor o igual $1$ y los valores de energía correspondientes son: $E_n=\frac{n²\pi²\hbar²}{2m_pl²}=\frac{n²h²}{8m_pl²}$, donde $n\in \{1,2,3,...\}$.

En la siguiente gráfica se representan las funciones de onda para los estados más bajos de energía.

```python

for n in list(range(1,6+1))+list(range(50,51)):
  FunOnda(n,1)
```

Primeros valores de energía

```python
#@title
# Primeros nmax valores de energía para el sistema SIN perturbar

z=np.linspace(0,l,200)

for n in range(1,6+1):
    E_n=(n**2*h_b**2*np.pi**2)/(2*m_p*l**2)
    print('n=',n,';','E=',E_n)
    plt.plot(z,E_n+(z-z),label=round(E_n,3))
    plt.title('Valores de Energía')
    plt.xlabel('Amplitud de la caja')
    plt.ylabel('Energía (Hartree)')
    plt.legend()

print('n=50 ; E=',(50**2*h_b**2*np.pi**2)/(2*m_p*l**2))
```

Graficos de $|\psi_n|²$ para los estados de energía más bajos.

```python
#@title
# Funciones de densidad de probabilidad para la partícula confinada

for n in list(range(1,6+1))+list(range(50,51)):
  fdpOnda(n,1)
```

En el gráfico de los cuadrados de las funciones de onda, es decir, de las funciones de densidad de probabilidad para encontrar la partícula en una región determinada entre $0$ y $l$, vemos que la partícula tiende a evitar los muros en los estados energéticos más bajos, pero a medida que $n$ incrementa la función de densidad tiende a ser uniformemente distribuida. La distribución para valores de $n$ grandes corresponde al comportamiento clásico de una partícula que permanece en cada punto entre los muros de la caja el mismo tiempo promedio. Este en un ejemplo del $\textbf{principio de correspondencia}$, el cual establece que la mecánica clásica emerge de la mecánica cuántica cuando los numéros cuánticos son muy grandes.
