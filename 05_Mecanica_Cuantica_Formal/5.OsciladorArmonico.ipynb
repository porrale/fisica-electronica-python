```python
import matplotlib.pyplot as plt   # Paquete gráfico de matplotlib 
from sympy import *               # Se importan todas las funciones del paquete sympy
init_printing()                   # Mejora la impresión visual de los resultados arrojados por sympy
import numpy as np                # Trae las funciones del paquete numpy llamandolas a través del prefijo np

x,y,z,alpha,omega=symbols('x,y,z,alpha,omega')            # Parámetros algebraicos para sympy

m_p = 1  # masa de la particula
h_b = 1  # hbar

#Función que genera los polinomios de Hermite

def H(z,n):
    if n==0:
        return 1
    if n==1:
        return 2*z
    if n > 1:
        return simplify(2*z*H(z,n-1)-2*(n-1)*H(z,n-2))


#Función que grafica los polinomios de Hermite

def Her(n):    
    fig, ax = plt.subplots()    
    z=np.linspace(-20,20,200)
    w=np.array(H(z,n))
    ax.plot(z,w)
    print ('n={:d}:'.format(n),'H_{:d}(x)='.format(n), H(x,n)) 

# Función que genera las expresiones de los osciladores armónicos

def psi(x,alpha,n):
    return (1/(sqrt((2**n)*factorial(n))))*((alpha/pi)**(1/4))*H(sqrt(alpha)*x,n)*exp((-alpha*x**2)/2)

def Psi(n):
    return psi(x,alpha,n)

# Funciones de onda para el oscilador armónico en donde entra en juego el parámetro alpha

def s(x,alpha,n):
  M=[]
  for i in range(len(x)):
    M.append(N(psi(x[i],alpha,n)))
  return M   

def PsiG(alpha,n):
  fig, ax = plt.subplots()
  x=np.linspace(-10,10,200)
  y=s(x,alpha,n)
  ax.plot(x,y,label='$\psi{:d}$'.format(n))
  ax.set_title('Función de onda oscilador armónico')
  ax.legend()
```

# **OSCILADOR ARMÓNICO UNIDIMENSIONAL**

La energía de una molécula puede aproximarse a través de la suma de las energías traslacional, rotacional, vibracional y electrónica. Los niveles de energía vibracionales más bajos de una molécula diatómica pueden aproximarse mediante los niveles de energía del oscilador armónico, ya que una parte de este sistema experimenta una fuerza de restauración proporcional al desplazamiento de los núcleos atómicos, con respecto a su centro de masas. La importancia del oscilador armónico radica también en el algebra involucrada en su solución, ya que esta aparece nuevamente en diferentes problemas, como por ejemplo, el tratamiento de movimientos rotacionales.

La fuerza asociada a un oscilador armónico unidimensional está dada por la ley de Hooke $F=-kx$, donde $k$ es la constante de restauración de la fuerza. Como la fuerza que actúa sobre la partícula es el negativo del gradiente de la energía potencial, tenemos que $F=-dV/dx$ y por lo tanto la energía potencial del sistema está dada por $V=\frac{kx^2}{2}$. Así el operador Hamiltoniano para el oscilador armónico asociado a una partícula de masa $m_p$, tiene el siguiente aspecto:

$$H=-\frac{\hbar^2}{2m_p}\frac{d^2}{dx^2}+\frac{kx^2}{2}$$

Definiendo la frecuencia angular $\omega$ del oscilador por $\omega^2=\frac{k}{m_p}$, el Hamiltoniano toma la forma $H=-\frac{\hbar^2}{2m_p}\frac{d^2}{dx^2}+\frac{m_p\omega^2x^2}{2}$. Las soluciones a la ecuación de $Schr\ddot{o}dinger$, $H\psi=E\psi$, para este sistema están dadas por 

$$\psi_n=\frac{1}{\sqrt{2^nn!}}(\frac{\alpha}{\pi})^{1/4}H_n(\sqrt{\alpha}x)e^{-\frac{\alpha x^2}{2}}, \ \ \ \alpha=\frac{m\omega}{\hbar}$$

donde $H_n(z)$ son los $\textbf{polinomios de Hermite}$. El valor de energía correspondientes para el estado $\psi_n$ es $E_n=\hbar\omega(n+1/2)$, donde $n\in \{0,1,2,3,...\}$.

## **VALORES DE ENERGÍA PARA EL OSCILADOR ARMÓNICO**

En el siguiente gráfico se muestran los valores de energía del oscilador armónico para los primeros ocho estados.

```python
# Gráfico de los valores de energía para los primeros estados del oscilador armónico
w=np.linspace(-5,5,100)

for n in range(0,8):
    En=h_b*(n+0.5)
    print('n=',n,';','E=',En)
    plt.plot(w,w**2/2)
    plt.plot(w,En+(w-w),label='$E_{:d}$'.format(n))
    plt.title('Valores de energía')
    plt.ylabel('Energía (Hartree)')
    plt.xlabel('x')
    plt.ylim(0,9)
    plt.legend()
```

Las características de una vibración de estiramiento atómica se pueden representar por un modelo mecánico que consta de dos masas unidas por un resorte. La perturbación de una de estas masas a lo largo del eje del resorte produce una vibración denominada $\textbf{movimiento armónico simple}$. Por lo regular, se plantea la aproximación de que el comportamiento de una vibración molecular es análogo al modelo mecánico que se acaba de describir. Por lo tanto, la frecuencia de la vibración molecular se calcula a través de los resultados obtenidos para el oscilador armónico.

Para una molécula diatómica que consta de dos masas $m_1$ y $m_2$, podemos representar su frecuencia vibracional por $\nu=\hbar\omega$, donde $\nu=\hbar\sqrt{\frac{k}{\mu}}$, siendo $\mu=\frac{m_1m_2}{m_1+m_2}$ (masa reducida del sistema). En este caso la cantidad $k$ se convierte en la constante de fuerza del enlace químico (medida de su rigidez). Por lo tanto, los niveles vibracionales de energía de una molécula diatómica están dados por

$$E_n=\hbar\sqrt{\frac{k}{\mu}}(n+1/2), \ donde \ \mu=\frac{m_1m_2}{m_1+m_2} \ y \ n=0,1,2,3,..$$

Mediante mediciones en la región del infrarrojo y la ecuación anterior es posible determinar las constantes de fuerza de distintos tipos de enlaces químicos. Por lo general, se ha encontrado que $k$ se encuentra entre $3*10^2$ y $8*10^2 \ N/m$ en la mayoría de los enlaces sencillos. El valor de $5*10^2$ se puede tomar como un valor promedio razonable. De esta misma forma se ha establecido que los enlaces dobles y triples tienen una constante de fuerza de dos a tres veces este valor ($1*10^3$ y $1.5*10^3 \ N/m$, respectivamente).
