```python
import numpy as np

m = 9.11e-31 # kg masa del electron
hbarra = 1.055e-34 # J s
L = .9e-9
raiz = np.sqrt(2 * m * eVDiferencia )
print(raiz)
C = raiz/hbarra
print(C)
exponente = -2 * C * L
exponente = round(exponente, 3) 
print(exponente)
T = np.exp(exponente) 
Tporcentual= 100 * T
print(T)
print(Tporcentual)
```
