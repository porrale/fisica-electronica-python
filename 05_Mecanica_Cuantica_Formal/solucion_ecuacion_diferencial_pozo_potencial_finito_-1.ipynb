```python
from sympy import * # otra manera de importar una Biblioteca completa

x, y, z = symbols('x y z')
init_printing(use_unicode=True)
f = symbols('f', cls=Function)   
init_printing(use_latex='mathjax')
```

```python
diffeq = Eq(f(x).diff(x, x)- f(x), 0)
```

```python
diffeq # imprime ecuacion diferencial
```

Solo vuelvo a escribir el resultado en LATEX.

$$- f{\left (x \right )} + \frac{d^{2}}{d x^{2}} f{\left (x \right )} = 0$$

```python
dsolve(diffeq, f(x))  #resuelve e imprime
```
