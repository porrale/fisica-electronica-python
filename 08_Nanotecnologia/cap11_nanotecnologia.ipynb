## 1. Límite de Abbe y Longitud de Onda de De Broglie en Microscopía Electrónica
El límite de resolución óptica de Abbe es:
$$R = \frac{\lambda}{2 \text{NA}}$$
Para un haz de electrones acelerado por un potencial $V$:
$$\lambda_e = \frac{h}{\sqrt{2 m_e e V}}$$

```python
import numpy as np
import matplotlib.pyplot as plt

h = 6.626e-34
me = 9.109e-31
e = 1.602e-19
c = 3.0e8

V_kV = np.linspace(1, 200, 200) # Tensión aceleradora de 1 a 200 kV
V = V_kV * 1e3

# De Broglie no relativista
lambda_e = h / np.sqrt(2 * me * e * V)

# De Broglie con corrección relativista: lambda = hc / sqrt(eV (eV + 2 me c^2))
lambda_e_rel = (h * c) / np.sqrt(e * V * (e * V + 2 * me * c**2))

plt.figure(figsize=(8, 5))
plt.plot(V_kV, lambda_e * 1e12, 'b--', label='Clásica no relativista')
plt.plot(V_kV, lambda_e_rel * 1e12, 'r-', lw=2.5, label='Corrección Relativista (TEM)')
plt.axhline(500, color='gray', ls=':', label='Luz visible (~500 nm)')
plt.axhline(13.5, color='orange', ls=':', label='Litografía EUV (13.5 nm)')
plt.yscale('log')
plt.title('Longitud de Onda de De Broglie del Electrón en Función de la Tensión')
plt.xlabel('Tensión de Aceleración $V$ [kV]')
plt.ylabel('$\lambda$ [picómetros (pm)]')
plt.grid(True, which='both', linestyle='--')
plt.legend()
plt.show()
```

## 2. Transición Planar a FinFET y GAAFET: Corrientes de Fuga Cuántica
A medida que la longitud de canal cae por debajo de $20\text{ nm}$, el efecto túnel a través del óxido y la fuga fuente-drenador degradan la pendiente subumbral ($SS$):
$$I_{\text{sub}} \propto 10^{(V_{GS} - V_{th})/SS}$$
Las arquitecturas 3D (**FinFET** y **GAAFET**) rodean el canal con la compuerta, reduciendo drásticamente la fuga en estado inactivo ($I_{\text{off}}$).

```python
vgs = np.linspace(0, 1.0, 300)
Vth = 0.35 # Tensión de umbral

# Pendientes subumbral típicas (mV/década):
# Planar: SS ~ 100 mV/dec (degradada por efectos de canal corto)
# FinFET: SS ~ 75 mV/dec
# GAAFET: SS ~ 65 mV/dec (cercana al límite térmico ideal de 60 mV/dec)
SS_planar = 0.100
SS_finfet = 0.075
SS_gaafet = 0.065

I_on = 1e-3 # 1 mA
i_planar = np.where(vgs < Vth, I_on * 10**((vgs - Vth)/SS_planar), I_on * (1 + (vgs - Vth)/0.1))
i_finfet = np.where(vgs < Vth, I_on * 10**((vgs - Vth)/SS_finfet), I_on * (1 + (vgs - Vth)/0.1))
i_gaafet = np.where(vgs < Vth, I_on * 10**((vgs - Vth)/SS_gaafet), I_on * (1 + (vgs - Vth)/0.1))

plt.figure(figsize=(8, 5))
plt.plot(vgs, i_planar, 'r--', lw=2, label='MOSFET Planar (SS = 100 mV/dec)')
plt.plot(vgs, i_finfet, 'b-.', lw=2, label='FinFET 3D (SS = 75 mV/dec)')
plt.plot(vgs, i_gaafet, 'g-', lw=2.5, label='GAAFET Nanoláminas (SS = 65 mV/dec)')
plt.yscale('log')
plt.title('Control Electrostático Subumbral: Planar vs FinFET vs GAAFET')
plt.xlabel('Tensión de Compuerta $V_{GS}$ [V]')
plt.ylabel('Corriente de Drenador $I_D$ [A] (Escala Logarítmica)')
plt.grid(True, which='both', linestyle='--')
plt.legend()
plt.show()
```
