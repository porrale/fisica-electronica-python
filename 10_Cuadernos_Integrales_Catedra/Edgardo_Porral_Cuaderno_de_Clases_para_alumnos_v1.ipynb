### Operadores lógicos

Los operadores lógicos en Python son muy explicitos:

    A == B  (A igual que B)
    A > B   (A mayor que B)
    A < B   (A menor que B)
    A >= B  (A igual o mayor que B)
    A <= B  (A igual o menor que B)
    A != B  (A diferente que B)
    A in B  (A incluido en B)
    A is B  (Identidad: A es el mismo elemento que B)

y a todos los podemos combinar con `not`, que niega la condición

````python
print ('¿20/3 == 6?',20/3 == 6)
print ('¿20//3 == 6?', 20//3 == 6)
print ('¿20//3 >= 6?', 20//3 >= 6)
print ('¿20//3 > 6?', 20//3 > 6)

print ('a == b:', a == b)
print ('a is b:',a is b)
print ('a es not b:',a is not b)
````

## FOR

Un bucle for establece la variable iteradora en cada valor de una lista, arreglo o cadena proporcionada y repite el código en el cuerpo del bucle for para cada valor de la variable iteradora.

En el ejemplo a continuación, usamos un bucle for para imprimir cada número de nuestro arreglo.


````python
# Ejemplo bucle for
for i in [1, 2, 3, 4]:
    print(i, end=", ") # prints: 1, 2, 3, 4,
````

````python
# Ejemplo más complejo
for i in [1, 3, 5, 7, 9]:
    x = i**2 - (i-1)*(i+1)
    print(x, end=", ") 
````

````python

# Ejemplo con un argumento
for i in range(5):
    print(i, end=", ") # 
````

La sintáxis para la creación de rangos es la siguiente:

  range(start, stop, step)

````python

import numpy as np # en este ejemplo estamos importanto NUMPY

import matplotlib.pyplot as plt # importamos Pyplot
from matplotlib.path import Path # Un módulo para tratar con las polilíneas 
# utilizadas en Matplotlib + INFO https://matplotlib.org/3.1.1/api/path_api.html
from matplotlib.patches import PathPatch

N = 400
t = np.linspace(0, 2 * np.pi, N) # preguntar ? 
r = 0.5 + np.cos(t) 
x, y = r * np.cos(t), r * np.sin(t)

fig, ax = plt.subplots()
ax.plot(x, y, "k")
ax.set(aspect=1)

def draw_error_band(ax, x, y, err, **kwargs):
    # Calcular normales a través de diferencias finitas centradas (excepto el primer punto
    # which uses a forward difference and the last point which uses a backward
    # diferencia).
    dx = np.concatenate([[x[1] - x[0]], x[2:] - x[:-2], [x[-1] - x[-2]]])
    dy = np.concatenate([[y[1] - y[0]], y[2:] - y[:-2], [y[-1] - y[-2]]])
    l = np.hypot(dx, dy)
    nx = dy / l
    ny = -dx / l

    # end points of errors
    xp = x + nx * err
    yp = y + ny * err
    xn = x - nx * err
    yn = y - ny * err

    vertices = np.block([[xp, xn[::-1]],
                         [yp, yn[::-1]]]).T
    codes = np.full(len(vertices), Path.LINETO)
    codes[0] = codes[len(xp)] = Path.MOVETO
    path = Path(vertices, codes)
    ax.add_patch(PathPatch(path, **kwargs))


axs = (plt.figure(constrained_layout=True)
       .subplots(1, 2, sharex=True, sharey=True))
errs = [
    (axs[0], "error constante", 0.05),
    (axs[1], " error variable", 0.05 * np.sin(2 * t) ** 2 + 0.04),
]
for i, (ax, title, err) in enumerate(errs):
    ax.set(title=title, aspect=1, xticks=[], yticks=[])
    ax.plot(x, y, "k")
    draw_error_band(ax, x, y, err=err,
                    facecolor=f"C{i}", edgecolor="none", alpha=.3)

plt.show()
````

estímulos o excitaciones del experimento y 𝑧 el resultado, la salida.

Ahora bien, pensándolo de esa manera, ¿cómo se hace para hallar la forma de la función (o sea, 𝑓)?
¿Y por qué es esto importante?
En primer lugar, es importante porque al conocer la forma de la función, logramos identificar una
ley de comportamiento de la naturaleza, tal que, al ajustar los parámetros de dicha función,
podemos controlar el resultado. Cabe destacar, para mejor comprensión del alumno, que suele
llamarse “parámetros” a aquellos valores de la función que no son las entradas ni las salidas (en el
caso de una recta, la pendiente y la ordenada).
Hay varias formas de intentar descubrir a esta función misteriosa:

* Con una función de ajuste. Es decir, hallando una función que aproxime a esos valores.
* Con un método de interpolación. Es decir, hallando una función que pase por todos los
valores medidos (o la composición de un conjunto de funciones, como funciones partidas).

Es importante remarcar que “ajustar” no es lo mismo que “interpolar”. Son 2 técnicas distintas y
más o menos útiles en función de la situación y las características del experimento. Cuando
ajustamos, creamos una función que aproxima sus valores a los valores experimentales (o sea, hay
una diferencia entre el valor medido -real- y el de ajuste) y cuando interpolamos creamos una
función que pasa por los valores medidos (no hay diferencia entre los valores de la función de
interpolación y los valores medidos -reales-).

#### Ejemplo

````python
import matplotlib.pyplot as plt
import numpy as np
# Data experimental
x = np.array([0.25, 0.5, 0.75, 1, 1.25, 1.5, 1.75, 2, 2.25, 2.5, 2.75, 3])
y = np.array([2.55, 0.65, -2.35, -6.2, -7.25, -10, -12.15, -15, -18.05, -18.7, -22, - 24.4])

# Impresion de los datos
plt.plot(x, y, 'o')
plt.xlabel('tiempo')
plt.ylabel('velocidad')
plt.grid()
plt.show()
````


Aquí es cuando recurrimos a la matemática, de manera de encontrar la forma de ajustar los puntos
a una recta. El método consiste en minimizar el error cuadrático medio (conocido como MSE, del
inglés “Mean Squared Error”), o lo que es lo mismo, encontrar la recta que genere la mínima
distancia a todos los puntos (de las infinitas rectas posibles)


````python
x_vals = x
def linefit(x, y):
# Devuelve la pendiente y la ordenada al origen del ajuste lineal al dataset
  x_prom = x.mean()
  pend = (y * (x - x_prom)).sum() / (x * (x - x_prom)).sum()
  yint = y.mean() - pend * x_prom
  return pend, yint
def abline(pend, yint, x, y):
  # Imprime una línea en base a una pendiente y una ordenada al origen
  y_vals = yint + pend * x_vals
  plt.plot(x, y, 'o', color = 'red')
  plt.plot(x, y_vals, '-')
pend,yint = linefit(x, y)
abline(pend, yint , x, y)
````

### Calidad del ajuste
A simple vista, parece que la recta hallada es muy buena, ¿pero existe alguna forma de validarlo?
Podemos usar el coeficiente de determinación (comúnmente llamado R2 o “R Cuadrado”).


````python
pend, yint = linefit(x,y)
def r2(pend, yint, x, y):
   # Devuelve el R^2 del ajuste lineal
   yhat = yint + pend * x
   ybar = y.mean()
   ssreg = np.sum((yhat - ybar) ** 2)
   sstot = np.sum((y - ybar) ** 2)
   return ssreg
r2(pend, yint,x,y )
````

````python
import numpy as np
import matplotlib.pyplot as plt
	
# Data experimental
x = np.array([0.25, 0.5, 0.75, 1, 1.25, 1.5, 1.75, 2, 2.25, 2.5, 2.75, 3])
y = np.array([2.55, 0.65, -2.35, -6.2, -7.25, -10, -12.15, -15, -18.05, -18.7, -22, -24.4])

# Devuelve la pendiente y la ordenada al origen del ajuste lineal al dataset
def linefit(x, y):
  x_prom = x.mean()
  pend = (y * (x - x_prom)).sum() / (x * (x - x_prom)).sum()
  yint = y.mean() - pend * x_prom
  return pend, yint
	
# Calcula el r-squared de la funcion
def r2(pend, yint, x, y):
  yhat = pend * x + yint
  ybar = y.mean()
  ssreg = np.sum((yhat - ybar) ** 2)
  sstot = np.sum((y - ybar) ** 2)
  return ssreg / sstot 
	
# Calcula los errores en los coeficientes
def err_parametros(pend, yint, x, y):
  n = len(x)
  sig = np.sqrt(np.sum((y - pend * x - yint) ** 2) / (n - 2))
  d_pend = (np.sqrt(n) * sig) / np.sqrt(n * np.sum(x ** 2) - (np.sum(x)) ** 2)
  d_yint = d_pend * np.sqrt(np.sum(x ** 2) / n)
  return d_pend, d_yint
	
	# Imprime la recta de ajuste sobre los valores medidos
def abline(pend, yint, x, y, d_pend, d_yint, r_2):
  print('Pendiente: ', pend)
  print('Ordenada: ', yint)
  print('Error en la pendiente: ', d_pend)
  print('Error en la ordenada: ', d_yint)
  print('Calidad del ajuste R^2: ', r_2)
	
# Crea el grafico
  y_vals = yint + pend * x
  plt.figure(1, figsize = (6, 4))
  plt.plot(x, y, 'o', color = 'red')
  plt.plot(x, y_vals, '--')
  plt.show()
	
pend, yint = linefit(x,y)
r_2 = r2(pend, yint, x, y)
d_pend, d_yint = err_parametros(pend, yint, x, y)
abline(pend, yint, x, y, d_pend, d_yint, r_2)
````

````python
import matplotlib.pyplot as plt    
import numpy as np    
        
# Data experimental    
x = np.array([0.25, 0.5, 0.75, 1, 1.25, 1.5, 1.75, 2, 2.25, 2.5, 2.75, 3])  
y = np.array([2.55, 0.65, -2.35, -6.2, -7.25, -10, -12.15, -15, -18.05, -18.7,-22, -24.4])  
dy = np.array([2, 3, 1, 5, 1, 2, 4, 1, 1, 5, 2, 1])  
  
def linefit_chi(x, y, dy):    
    # Devuelve la pendiente y la ordenada al origen del ajuste lineal al dataset    
    x_prom = (x / (dy**2)).sum() / (1 / (dy**2)).sum()    
    y_prom = (y / (dy**2)).sum() / (1 / (dy**2)).sum()   
    pend = (y / (dy**2) * (x - x_prom)).sum() / (x / (dy**2) * (x - x_prom)).sum()    
    yint = y_prom - pend * x_prom    
    return pend, yint    
  
def abline_chi(pend, yint, x, y, dy):    
    # imprime una línea en base a una pendiente y una ordenada al origen       
    y_vals = yint + pend * x  
  
    # crea el grafico    
    plt.figure(1, figsize = (6,4))  
    plt.plot(x, y, 'o', color = 'red')    
    plt.plot(x, y_vals, '--')    
    plt.errorbar(x, y, fmt = 'ro', xerr = 0, yerr = dy, ecolor = 'black')    
    plt.show()   
  
pend, yint = linefit_chi(x, y, dy)  
print("Pendiente: ", pend)  
print("Ordenada: ", yint)  
abline_chi(pend, yint, x, y, dy)  
def err_parametros(pend, yint, x, y, dy):
  x_prom = (x / (dy**2)).sum() / (1 / (dy**2)).sum()
  d_pend = np.sqrt(1 / ((x - x_prom) * (x / (dy**2))).sum())
  d_yint = np.sqrt(d_pend * (np.sum((x**2) / (dy**2)) / np.sum(1 / (dy**2))))
  return d_pend, d_yint
````

##Función de Masa de Probabilidad
Otra forma de representar a las distribuciones discretas es utilizando su Función de Masa de Probabilidad o FMP, la cual relaciona cada valor con su *probabilidad* en lugar de su frecuencia como vimos anteriormente. Esta función es normalizada de forma tal que el valor total de *probabilidad* sea 
1. La ventaja que nos ofrece utilizar la FMP es que podemos comparar dos distribuciones sin necesidad de ser confundidos por las diferencias en el tamaño de las *muestras. También debemos tener en cuenta que FMP funciona bien si el número de valores es pequeño; pero a medida que el número de valores aumenta, la *probabilidad asociada a cada valor se hace cada vez más pequeña y el efecto del ruido aleatorio aumenta. Veamos un ejemplo con Python.

````python
# <!-- collapse=True -->
# importando modulos necesarios
%matplotlib inline

import matplotlib.pyplot as plt
import numpy as np 
from scipy import stats 
import seaborn as sns 

# Graficando FMP
n, p = 30, 0.4 # parametros de forma de la distribución binomial
n_1, p_1 = 20, 0.3 # parametros de forma de la distribución binomial
x = np.arange(stats.binom.ppf(0.01, n, p),              stats.binom.ppf(0.99, n, p))
x_1 = np.arange(stats.binom.ppf(0.01, n_1, p_1),              stats.binom.ppf(0.99, n_1, p_1))
fmp = stats.binom.pmf(x, n, p) # Función de Masa de Probabilidad
fmp_1 = stats.binom.pmf(x_1, n_1, p_1) # Función de Masa de Probabilidad
plt.plot(x, fmp, '--')
plt.plot(x_1, fmp_1)
plt.vlines(x, 0, fmp, colors='b', lw=5, alpha=0.5)
plt.vlines(x_1, 0, fmp_1, colors='g', lw=5, alpha=0.5)
plt.title('Función de Masa de Probabilidad')
plt.ylabel('probabilidad')
plt.xlabel('valores')
plt.show()
````
